```python
import sqlite3

conn = sqlite3.connect('employee.db') # it create the file if doesn't exist, and if exist just connect it

# create cursor
c = conn.cursor()

# now we can create table and it's columns using c and execute method, and write the SQL command inside
c.execute("""CREATE TABLE employees (
            first text,
            last text,
            pay integer
            )""")

conn.commit()
```

```python
# Now, we want to insert some data into employees table

c.execute("INSERT INTO employees VALUES ('Maryam', 'Skaik', 90000)")
conn.commit()
```

```python
# create functions for insert, get, update and delete employee
def insert_emp(emp):
    with conn:
        c.execute("INSERT INTO employees VALUES (?, ?, ?)", (emp.first, emp.last, emp.pay))

def get_emps_by_name(lastname):
    c.execute("SELECT * FROM employees WHERE last=:last", {'last': lastname})
    return c.fetchall()

def update_pay(emp, pay):
    with conn:
        c.execute(""" UPDATE employees SET pay =:pay
                    WHERE first = :first AND last = :last""",
                    {'first': emp.first, 'last': emp.last, 'pay': pay})
        emp.pay = pay

def remove_emp(emp):
    with conn:
        c.execute("Delete from employees Where first = :first AND last = :last",
                     {'first': emp.first, 'last': emp.last})
```

```python
insert_emp(Employee("Yomna", "Abdo", 2000))
insert_emp(Employee("Sali", "Soso", 5000))
```
