```python
# class
class Employee:

    # variables
    raise_amount = 1.04
    num_of_emps = 0
    
    # use constructor (init)
    def __init__(self, first, last, pay):
        self.first = first
        self.last = last
        #self.email = first + '.' + last + '@compay.com'
        self.pay = pay
        Employee.num_of_emps += 1

    def full_name(self):
        return f'{self.first} {self.last}'

    def apply_raise(self):
        self.pay = int(self.pay * self.raise_amount)

    @classmethod
    def set_raise_amt(cls, amount):
        cls.raise_amount = amount

    # Magic methods
    def __repr__(self): # will work when we call the instance or print it
        return f'Employee ({self.first}, {self.last}, {self.pay})'

    def __str__(self):
        return f'{self.first} {self.last} - {self.email})'

    def __add__(self, other):
        return self.pay + other.pay

    # Getters and Setters
    @property
    def fullname(self):
        return f'{self.first} {self.last}'

    @property
    def email(self):
        return f'{self.first}.{self.last}@compay.com'

    @fullname.setter
    def fullname(self, name):
        first, last = name.split(' ')
        self.first = first
        self.last = last

    @fullname.deleter
    def fullname(self):
        print('Delete Name!')
        self.first = None
        self.last = None
        
```

```python
# subclasses
class Developer(Employee):
    raise_amount = 1.1

    def __init__(self, first, last, pay, prog_lang):
        super().__init__(first, last, pay)
        self.prog_lang = prog_lang
```

```python
class Maneger(Employee):
    
    def __init__(self, first, last, pay, employees=None):
        super().__init__(first, last, pay)
        if employees is None:
            self.employees = []
        else:
            self.employees = employees

    def add_emp(self, emp):
        if emp not in self.employees:
            self.employees.append(emp)

    def remove_emp(self, emp):
        if emp in self.employees:
            self.employees.remove(emp)

    def print_emps(self):
        for emp in self.employees:
            print(emp.__dict__)
```

```python
# instance

e1 = Employee('Maryam', 'Skaik', 90000)
e2 = Employee('Tia', 'lulu', 60000)

print(e1.email)
print(e2.email)
```
