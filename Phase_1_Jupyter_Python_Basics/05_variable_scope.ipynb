# Variable Scope
## LEGB
1. Local: variables defined within an function
2. Enclosing: variables in local scope of enclosing function
3. Global: variables defined in the top level of a module, or declared Global
4. Built-in: names that preassigned in Python

```python
x = 'global x'

def test():
    y = 'local y'
    print(y)
    print(x)
    
test()
```

```python
def test():
    x = 'local x'
    print(x)

test()
x
```

```python
def test():
    global x
    x = 'local x'
    print(x)

test()
x
```
