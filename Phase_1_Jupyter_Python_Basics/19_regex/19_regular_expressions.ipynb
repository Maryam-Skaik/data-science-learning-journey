# re Module - How to Write and Match Regular Expressions (Regex)

```python
import re

text_to_search = '''
abcdefghijklmnopqurtuvwxyz
ABCDEFGHIJKLMNOPQRSTUVWXYZ
1234567890

Ha HaHa

MetaCharacters (Need to be escaped):
. ^ $ * + ? { } [ ]  | ( )

Maryam.com

321-555-4321
123.555.1234
123*555*1234
800-555-1234
900-555-1234

Mr. Schafer
Mr Smith
Ms Davis
Mrs. Robinson
Mr. T
'''

sentence = 'Start a sentence and then bring it to an end'

# we will use compile method from re module, and specify the pattern

pattern = re.compile(r'abc')

matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
print(text_to_search[1:4])
```

```python
pattern = re.compile(r'.') # assume all matches equal . except new line

matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
# to search just about .
pattern = re.compile(r'\.') 

matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
# to search just about .
pattern = re.compile(r'Maryam\.com') 

matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
pattern = re.compile(r'\d') # all digits

matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
pattern = re.compile(r'\bHa') 

matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```

```python
pattern = re.compile(r'\BHa') 

matches = pattern.finditer(text_to_search)

for match in matches:
    print(match)
```
