# Generate Random Numbers and Data Using the random Module

```python
import random

value = random.random()
value # 0.40749809213147903, another run: 0.8557250052490754
# and so on, values between 0 and 1
```

```python
# if we want a random number in a spesific range, we use uniform 
value = random.uniform(1, 10)
value # 3.9243858203100506, 8.11192967671602
```

```python
# if we want just integer, we use randint
value = random.randint(1, 10)
value # 6, 10, 1, 5, ....
```

```python
greetings = ['Hello', 'Hi', 'Welcome', 'Hey']

value = random.choice(greetings)
value + ', Maryam' # 'Welcome, Maryam', 'Hey, Maryam', ''Hi, Maryam'
```

```python
colors = ['red', 'black', 'blue']

results = random.choices(colors, k=10)
results # ['black', 'blue', 'blue', 'blue', 'red', 'blue', 'blue', 'red', 'red', 'red'], ['red', 'black', 'blue', 'red', 'red', 'blue', 'blue', 'red', 'black', 'black']
```

```python
colors = ['red', 'black', 'blue']

results = random.choices(colors, weights= [15,3,31], k=10)
results # ['red', 'red', 'red', 'blue', 'red', 'blue', 'blue', 'blue', 'blue', 'blue'], ['blue', 'red', 'blue', 'red', 'blue', 'red', 'red', 'blue', 'blue', 'blue']
```

```python
deck = list(range(1, 53))
random.shuffle(deck)
deck
```

```python
deck = list(range(1, 53))
hand = random.sample(deck, k = 5)  
hand # [38, 48, 51, 24, 37], [12, 41, 30, 20, 34], [52, 42, 46, 45, 43], ....
```
