# Using Try/Except Blocks for Error Handling

```python
# structure

try:
    pass
except Exception:
    pass
else:
    pass
finally:
    pass

# we can only use try/except
```

```python
# if we want to open file is not exist, we have to use try/except to prevent crash

try:
    f = open('testfile.txt')
except Exception:
    print("This file isn't exist ):")
```

```python
# now, create the file, but add another exception
try:
    f = open('testfile.txt')
    var = my_var
except Exception:
    print("This file isn't exist ):")

# our exception is genral, so will catch all exceptions
```

```python
# try to specify exception

try:
    f = open('testfile.txt')
    var = my_var
except FileNotFoundError:
    print("This file isn't exist ):")

# as expected, we got error
```

```python
# solution is with add another catch with general exception
try:
    f = open('testfile.txt')
    var = my_var
except FileNotFoundError:
    print("This file isn't exist ):")
except Exception:
    print('Check your code')
    
```

```python
# to got a message for the error:
try:
    f = open('testfile.txt')
    var = my_var
except FileNotFoundError:
    print("This file isn't exist ):")
except Exception as e:
    print(e)
```

```python
# now we add else, which will executed if try didnn't throw an exception
try:
    f = open('testfile.txt')
except Exception:
    print("This file isn't exist ):")
else:
    print(f.read())
    f.close()
```

```python
# now finally executed if try throw an exception or if the code is work
try:
    f = open('testfile.txt')
except Exception:
    print("This file isn't exist ):")
else:
    print(f.read())
    f.close()
finally:
    print('Finish your code')
```

```python
# we can throw exception from try, using raise
try:
    f = open('testfile.txt')
    if f.name == 'testfile.txt':
        raise Exception
except Exception:
    print("This file isn't exist ):")
else:
    print(f.read())
    f.close()
finally:
    print('Finish your code')
```
