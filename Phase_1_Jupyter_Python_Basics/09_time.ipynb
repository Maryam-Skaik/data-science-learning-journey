```python
import datetime

d = datetime.date(2026, 1, 31)
d # 2026-01-31

today = datetime.date.today()
today # 2026-01-31
today.year # 2026

today.weekday() # 5, Monday -> 0, Sunday -> 6
today.isoweekday() # 6, Monday -> 1, Sunday -> 7

tdelta = datetime.timedelta(days=7)
print(today + tdelta) # 2026-02-07
print(today - tdelta) # 2026-01-24
```

```python
t = datetime.time(18, 26, 30, 1000)
print(t) # 18:26:30.001000
t.hour # 18
```
