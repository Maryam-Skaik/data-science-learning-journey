# File Objects

```python
# open .txt file
f = open('test.txt', 'r')
# r -> read file
# w -> write file
# a -> append
# r+ -> read and write

f.name # 'test.txt'

f.mode # 'r', mean we open it for reading

f.close()
```

```python
with open('test.txt', 'r') as f:
    pass
    
f.closed # True
```

```python
with open('test.txt', 'r') as f:
    f_contents = f.read()
    print(f_contents)
```

```python
with open('test.txt', 'r') as f:
    f_contents = f.readlines()
    print(f_contents)
```

```python
with open('test.txt', 'r') as f:
    f_contents = f.readline()
    print(f_contents)
```

```python
with open('test.txt', 'r') as f:
    f_contents = f.readline()
    print(f_contents, end='')
    
    f_contents = f.readline()
    print(f_contents, end='')
```

```python
with open('test.txt', 'r') as f:
    for line in f:
        print(line, end='')
```

```python
with open('test.txt', 'r') as f:
    f_contents = f.read(95)
    print(f_contents, end = '')
```

```python
with open('test.txt', 'r') as f:
    size_to_read = 10
    f_contents = f.read(size_to_read)
    while len(f_contents) > 0:
        print(f_contents, end = '$')
        f_contents = f.read(size_to_read)
```

```python
with open('test.txt', 'r') as f:
    size_to_read = 20
    f_contents = f.read(size_to_read)
    print(f.tell()) # return the current position of the file pointer
```

```python
with open('test.txt', 'r') as f:
    size_to_read = 20
    f_contents = f.read(size_to_read)
    print(f_contents, end = '')
    
    f.seek(0) # change the position of the file pointer

    f_contents = f.read(size_to_read)
    print(f_contents)
    
```

```python
with open('test2.txt', 'w') as f:
    f.write('Maryam')

# if the file exist, it will overwrite, if not, it will create it

# result: Maryam
```

```python
with open('test2.txt', 'w') as f:
    f.write('Maryam')
    f.write('Maryam')

# result: MaryamMaryam
```

```python
with open('test2.txt', 'w') as f:
    f.write('Maryam')
    f.seek(0)
    f.write('Maryam')

# result: Maryam
```

```python
with open('test2.txt', 'w') as f:
    f.write('Maryam')
    f.seek(0)
    f.write('S')

# result: Saryam
```

```python
# copy content from one file to another
with open('test.txt', 'r') as rf:
    with open('copy_test.txt', 'w') as wf:
        for line in rf:
            wf.write(line)
```

```python
 # copy image from one file to another, we have to use binary mode
with open('BMW.jpg', 'rb') as rf:
    with open('copy_BMW.jpg', 'wb') as wf:
        for line in rf:
            wf.write(line)
```

```python
 # copy image from one file to another, we have to use binary mode
with open('BMW.jpg', 'rb') as rf:
    with open('copy_BMW.jpg', 'wb') as wf:
        chunk_size = 4024
        rf_chunk = rf.read(chunk_size)
        while len(rf_chunk) > 0:
            wf.write(rf_chunk)
            rf_chunk = rf.read(chunk_size)
```
