# Automate Parsing and Renaming of Multiple Files

```python
import os

os.chdir('path')

os.getcwd() # to check my current folder
```

```python
for f in os.listdir():
    print(os.path.splitext(f)) # give first element = name, and second = extension
```

```python
for f in os.listdir():
    file_name, file_ext = os.path.splitext(f)
    print(file_name)
```

```python
for f in os.listdir():
    file_name, file_ext = os.path.splitext(f)
    print(file_name.split('#'))
```

```python
for f in os.listdir():
    file_name, file_ext = os.path.splitext(f)
    if '#' not in file_name:
        continue
    file_title, file_number = file_name.split('#')
    print(file_number)
```

```python
for f in os.listdir():
    file_name, file_ext = os.path.splitext(f)
    if '#' not in file_name:
        continue
    file_title, file_number = file_name.split('#')

    file_title = file_title.strip()
    file_number = file_number.strip()
    
    print(f'{file_number}_{file_title}{file_ext}')
```

```python
for f in os.listdir():
    file_name, file_ext = os.path.splitext(f)
    if '#' not in file_name:
        continue
    file_title, file_number = file_name.split('#')

    file_title = file_title.strip()
    file_number = file_number.strip().zfill(2) # make number contain 2 digits
    
    print(f'{file_number}_{file_title}{file_ext}')
```

```python
# now we can rename file
for f in os.listdir():
    file_name, file_ext = os.path.splitext(f)
    if '#' not in file_name:
        continue
    file_title, file_number = file_name.split('#')

    file_title = file_title.strip()
    file_number = file_number.strip().zfill(2) # make number contain 2 digits
    
    new_name = f'{file_number}_{file_title}{file_ext}'

    os.rename(f, new_name)
```
