# Working with JSON Data using the json Module

```python
import json

people_string = '''
{
    "people": [
        {
            "name": "Maryam Skaik",
            "phone": "222-555-1536",
            "emails": ["maryam@personalemail.com", "maryam@company.com"],
            "has_license": false
        },
        {
            "name": "Tia lulu",
            "phone": "333-555-1578",
            "emails": null,
            "has_license": true
        }
    ]
}
'''

data = json.loads(people_string)

print(type(data)) # <class 'dict'>

print(data)
```

```python
print(type(data["people"])) # <class 'list'>
```

```python
for person in data['people']:
    print(person['name'])
```

```python
# we can also convert it to string
new_string = json.dumps(data)
new_string
```

## read json data from file

```python
with open('states.json', 'r') as f:
    data = json.load(f)
```

```python
for state in data['states']:
    print(state)
```

```python
for state in data['states']:
    del state['area_codes']
```

```python
with open('states.json', 'w') as f:
    json.dump(data, f, indent=2)
```

## From API

```python
from urllib.request import Request, urlopen

url = "https://liveapi.in/geo/state/?country=US&format=json"

req = Request(
    url,
    headers={
        "User-Agent": "Mozilla/5.0"
    }
)

with urlopen(req) as response:
    source = response.read()

print(source)
```

```python
data = json.loads(source)
data
```
