```python
# Functions

def hello():
    pass

hello() # None

def hello():
    'Hello from function'

hello() # Hello from function

def hello(name):
    f'Hello, {name}'

hello('Maryam') # Hello, Maryam

# we can put a default value for argument
def hello(greeting, name='You'):
    f'{greeting}, {name}'

hello('Hi') # Hi, You

def student_info(*args, **kwargs):
    print(args)
    print(kwargs)

# student_info('DS', 'WD', name='Maryam', age=22)
# ('DS', 'WD')
# {'name': 'Maryam', 'age': 22}

courses = ('DS', 'WD')
info = {'name': 'Maryam', 'age': 22}

student_info(courses, info)
# (('DS', 'WD'), {'name': 'Maryam', 'age': 22})
# {}

# it treat with courses and info as args

student_info(*courses, **info)

# ('DS', 'WD')
# {'name': 'Maryam', 'age': 22}
```
