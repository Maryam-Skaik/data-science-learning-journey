# Sorting

```python
# Lists
li = [9, 1, 8, 2, 7, 3, 6, 5, 4]
s_li = sorted(li)
s_li # [1, 2, 3, 4, 5, 6, 7, 8, 9]

li # [9, 1, 8, 2, 7, 3, 6, 5, 4]

li.sort()
li # [1, 2, 3, 4, 5, 6, 7, 8, 9]

s_li = sorted(li, reverse=True)
s_li # [9, 8, 7, 6, 5, 4, 3, 2, 1]
```

```python
# Tuples
tup = (9, 1, 8, 2, 7, 3, 6, 5, 4)
s_tup = sorted(tup)
s_tup # [1, 2, 3, 4, 5, 6, 7, 8, 9], list

tup # (9, 1, 8, 2, 7, 3, 6, 5, 4)
```

```python
# Dictionary
di = {'name': 'Maryam', 'age': 22, 'job': 'TA'}
s_di = sorted(di)
s_di # ['age', 'job', 'name']
```

```python
li = [-6, -5, -4, 1, 2, 3]
# sort based on abs values
s_li = sorted(li, key=abs)
s_li # [1, 2, 3, -4, -5, -6]
```

```python
# Object
from operator import attrgetter
class Employee():
    def __init__(self, name, age, salary):
        self.name = name
        self.age = age
        self.salary = salary

    def __repr__(self):
        return f'{self.name}, {self.age}, ${self.salary}'

e1 = Employee('Maryam', 22, 90000)
e2 = Employee('Tia', 25, 70000)
e3 = Employee('Tima', 25, 60000)

employees = [e1, e2, e3]

def e_sort(emp):
    return emp.salary
    
s_employees = sorted(employees, key=e_sort)
s_employees # [Tima, 25, $60000, Tia, 25, $70000, Maryam, 22, $90000]

s_employees = sorted(employees, key=e_sort, reverse=True)
s_employees # [Maryam, 22, $90000, Tia, 25, $70000, Tima, 25, $60000]


s_employees = sorted(employees, key=lambda e:e.salary)
s_employees # [Tima, 25, $60000, Tia, 25, $70000, Maryam, 22, $90000]


s_employees = sorted(employees, key=attrgetter('age'))
s_employees # [Maryam, 22, $90000, Tia, 25, $70000, Tima, 25, $60000]
```
