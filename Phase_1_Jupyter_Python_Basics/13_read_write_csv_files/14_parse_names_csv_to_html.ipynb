# Parsing Names From a CSV to an HTML List

```python
import csv

html_output = ''

names = []

with open('employees.csv', 'r') as data_file:
    csv_data = csv.reader(data_file)

    # we don't want headers
    next(csv_data)

    for line in csv_data:
        names.append(f'{line[0]} {line[1]}')

for name in names:
    print(name)
```

```python
# construct the html output

html_output += f'<p>There are currently {len(names)} public contribtors. Thank you! </p>\n'
html_output
```

```python
html_output += '<ul>\n' 

for name in names:
    html_output += f'\t<li>{name}</li>\n'

html_output += '</ul>' 
print(html_output)
```

```python
# use dictionary

with open('employees.csv', 'r') as data_file:
    csv_data = csv.DictReader(data_file)

    for line in csv_data:
        names.append(f'{line['First Name']} {line['Last Name']}')

for name in names:
    print(name)
```

```python
# construct the html output

html_output += f'<p>There are currently {len(names)} public contribtors. Thank you! </p>\n'

html_output += '<ul>\n' 

for name in names:
    html_output += f'\t<li>{name}</li>\n'

html_output += '</ul>' 
print(html_output)
```

```python
# write names into html file
with open('names.html', 'w') as f:
    f.write(html_output)
```
