```python
import csv

with open('users.csv', 'r') as csv_file:
    csv_reader = csv.reader(csv_file)

    for line in csv_reader:
        print(line)
```

```python
with open('users.csv', 'r') as csv_file:
    csv_reader = csv.reader(csv_file)

    for line in csv_reader:
        print(line[2])
```

```python
# if we want to skip first line which contains columns names, we can use next function

with open('users.csv', 'r') as csv_file:
    csv_reader = csv.reader(csv_file)

    next(csv_reader)
    
    for line in csv_reader:
        print(line)
```

```python
# if we can write data from this .csv file to another
with open('users.csv', 'r', newline='') as csv_file:
    csv_reader = csv.reader(csv_file)

    with open('new_users.csv', 'w', newline='') as new_file: # add newline = '', Windows: avoid extra blank lines when writing CSV
        csv_writer = csv.writer(new_file, delimiter='\t') # tab delimiter

        for line in csv_reader:
            csv_writer.writerow(line)        
```

```python
# read new file
with open ('new_users.csv', 'r', newline='') as csv_file: # Windows: prevents empty rows when reading CSV
    csv_reader = csv.reader(csv_file, delimiter='\t')

    for line in csv_reader:
        print(line)
```

### Notes:
- newline= ' ' required on Windows to prevent extra blank lines
- fix for Windows CSV newline handling
- prevents csv module from inserting extra empty rows on Windows

```python
# we can convert csv file into dictionary

with open('users.csv', 'r') as csv_file:
    csv_reader = csv.DictReader(csv_file)

    for line in csv_reader:
        print(line)
```

```python
with open('users.csv', 'r') as csv_file:
    csv_reader = csv.DictReader(csv_file)

    for line in csv_reader:
        print(line['email'])
```

```python
with open('users.csv', 'r') as csv_file:
    csv_reader = csv.DictReader(csv_file)

    with open('new_usres2.csv', 'w') as new_file:
        fields_names = ['name', 'email', 'streetAddress', 'city', 'state', 'zipCode']

        csv_writer = csv.DictWriter(new_file, fieldnames=fields_names, delimiter='\t')

        csv_writer.writeheader()
        
        for line in csv_reader:
            csv_writer.writerow(line)
```

```python
with open('users.csv', 'r') as csv_file:
    csv_reader = csv.DictReader(csv_file)

    with open('new_usres2.csv', 'w') as new_file:
        fields_names = ['name', 'email', 'streetAddress', 'city', 'state']

        csv_writer = csv.DictWriter(new_file, fieldnames=fields_names, delimiter='\t')

        csv_writer.writeheader()
        
        for line in csv_reader:
            del line['zipCode']
            csv_writer.writerow(line)
```
