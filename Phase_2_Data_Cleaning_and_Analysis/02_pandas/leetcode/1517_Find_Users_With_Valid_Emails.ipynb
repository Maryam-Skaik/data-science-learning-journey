# 1517. Find Users With Valid E-Mails

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – String Matching & Regular Expressions  

🔗 Problem Link: https://leetcode.com/problems/find-users-with-valid-emails/

## 📝 Problem Description

We are given a Pandas DataFrame `users` with columns:

- user_id
- name
- mail

A valid e-mail must satisfy:

1. Prefix (before @) starts with a letter.
2. Prefix may contain letters, digits, `_`, `.`, or `-`.
3. Domain must be exactly `@leetcode.com`.

Return all users with valid e-mails.

## 💡 Key Concepts

1. Use **regex** to validate emails:
   - `^[A-Za-z]` → starts with a letter
   - `[A-Za-z0-9_.-]*` → allowed characters in prefix
   - `@leetcode\.com$` → domain must match exactly

2. Filter DataFrame rows using `.str.match(pattern)`  
3. Keep all columns of users satisfying the regex.

```python
import pandas as pd

def valid_emails(users: pd.DataFrame) -> pd.DataFrame:
    pattern = r'^[A-Za-z][A-Za-z0-9_.-]*@leetcode\.com$'
    valid_users = users[users['mail'].str.match(pattern)]
    return valid_users
```

## 🔎 Solution Explanation

Step 1: Define regex pattern for valid e-mail.  
Step 2: Use `.str.match(pattern)` to check each email.  
Step 3: Filter the rows of `users` matching the pattern.  

Time Complexity: O(n) – checking regex for each row  
Space Complexity: O(n) – filtered DataFrame

```python
data = {
    "user_id": [1,2,3,4,5,6,7],
    "name": ["Winston","Jonathan","Annabelle","Sally","Marwan","David","Shapiro"],
    "mail": [
        "winston@leetcode.com",
        "jonathanisgreat",
        "bella-@leetcode.com",
        "sally.come@leetcode.com",
        "quarz#2020@leetcode.com",
        "david69@gmail.com",
        ".shapo@leetcode.com"
    ]
}

users = pd.DataFrame(data)
valid_emails(users)
```
