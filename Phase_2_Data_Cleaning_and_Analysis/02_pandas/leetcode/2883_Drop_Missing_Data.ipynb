## 📝 Problem Description

We are given a Pandas DataFrame called `students`.

Columns:
- student_id
- name
- age

Some rows have missing values in the `name` column.

Our task is to:
- Remove the rows that have missing values in the `name` column.
- Return the cleaned DataFrame.


## 💡 Key Concept

In Pandas, missing data can be handled using:

```python
df.dropna()
```

To drop rows with missing values in specific columns, use:

```python
df.dropna(subset=['column_name'])
```

By default:

- It removes any row with missing data in the specified column(s).
- `inplace=True` modifies the DataFrame directly.

````python
import pandas as pd

def dropMissingData(students: pd.DataFrame) -> pd.DataFrame:
    students.dropna(subset=['name'], inplace=True)
    return students
````

## 🔎 Solution Explanation

Step 1: Use `dropna()` to remove rows with missing values.

```python
students.dropna(subset=['name'], inplace=True)
```

- `subset=['name']` → Removes rows where the `name` column has missing values.
- `inplace=True` → Modifies the DataFrame directly.

Step 2: Return the cleaned DataFrame.

Time Complexity: O(n), Pandas scans the name column once.

Space Complexity: O(1), No new DataFrame is created because we use inplace=True.

````python
data = {
    "student_id": [32, 217, 779, 849],
    "name": ["Piper", None, "Georgia", "Willow"],
    "age": [5, 19, 20, 14]
}

students = pd.DataFrame(data)

dropMissingData(students)
````
