## 📝 Problem Description

We are given a Pandas DataFrame `views` with columns:

- article_id
- author_id
- viewer_id
- view_date

Task: Find all authors who viewed **at least one of their own articles**.  

Return a DataFrame with one column:  
- id (author_id)  
- Remove duplicates  
- Sort the results in ascending order

## 💡 Key Concept

1. Filter rows where `author_id == viewer_id`.
2. Select only the author column.
3. Remove duplicates.
4. Sort results ascending.

````python
import pandas as pd

def article_views(views: pd.DataFrame) -> pd.DataFrame:
    # Filter authors who viewed their own article
    filtered = views[views["author_id"] == views["viewer_id"]]
    
    # Keep only author_id column and remove duplicates
    result = filtered[["author_id"]].drop_duplicates()
    
    # Rename column and sort
    result = result.rename(columns={"author_id": "id"}).sort_values(by="id")
    
    return result
````

## 🔎 Solution Explanation

1. **Filter rows** where authors viewed their own article:

```python
filtered = views[views["author_id"] == views["viewer_id"]]
```

2. Keep only author_id column and remove duplicates:

```python
result = filtered[["author_id"]].drop_duplicates()
```

3. Rename and sort:

```python
result.rename(columns={"author_id": "id"}).sort_values(by="id")
```

Time Complexity: O(n)

- `n` = number of rows in `views`

Space Complexity: O(n)

````python
data = {
    "article_id": [1, 1, 2, 2, 4, 3, 3],
    "author_id": [3, 3, 7, 7, 7, 4, 4],
    "viewer_id": [5, 6, 7, 6, 1, 4, 4],
    "view_date": [
        "2019-08-01", "2019-08-02", "2019-08-01",
        "2019-08-02", "2019-07-22", "2019-07-21",
        "2019-07-21"
    ]
}

views = pd.DataFrame(data)

article_views(views)
````
