## 📝 Problem Description

We are given a Pandas DataFrame called `students`.

Columns:
- student_id (int)
- name (object)
- age (int)
- grade (float)

There is an issue:
The `grade` column is stored as float values.

Our task is to:
- Convert the `grade` column to integer type.
- Return the updated DataFrame.

## 💡 Key Concept

In Pandas, we can change the data type of a column using:

```python
df['column'].astype(new_type)
```

Example:

```python
df['grade'] = df['grade'].astype(int)
```

This converts all values in the column to integers.

````python
import pandas as pd

def changeDatatype(students: pd.DataFrame) -> pd.DataFrame:
    students['grade'] = students.grade.astype(int)
    return students
````

## 🔎 Solution Explanation

Step 1: Access the `grade` column:

```python
students['grade']
```

Step 2: Convert its data type using:

```python
astype(int)
```

This converts each float value (e.g., 73.0) into an integer (73).

Step 3: Assign the converted column back to the DataFrame.

Step 4: Return the updated DataFrame.

Time Complexity: O(n), Each value in the `grade` column is converted once.

Space Complexity: O(n), A new integer column is created during conversion.

````python
data = {
    "student_id": [1, 2],
    "name": ["Ava", "Kate"],
    "age": [6, 15],
    "grade": [73.0, 87.0]
}

students = pd.DataFrame(data)

changeDatatype(students)
````
