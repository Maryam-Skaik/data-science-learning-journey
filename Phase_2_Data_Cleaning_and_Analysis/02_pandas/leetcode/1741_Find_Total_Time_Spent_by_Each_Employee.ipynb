# 1741. Find Total Time Spent by Each Employee

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – Grouping / Aggregation  

🔗 Problem Link: https://leetcode.com/problems/find-total-time-spent-by-each-employee/

## 📝 Problem Description

Table: `Employees`

Columns:
- emp_id (employee ID)
- event_day (day of the event)
- in_time (minute employee entered)
- out_time (minute employee left)

**Task:**

Calculate the total time spent by each employee on each day in the office. If an employee enters and leaves multiple times during a day, the total time spent that day is the sum of the durations for each entry.  

The time spent in the office for a single entry is `out_time - in_time`.

Return the result table in any order.

**Example:**

Input:

| emp_id | event_day  | in_time | out_time |
|--------|------------|---------|----------|
| 1      | 2020-11-28 | 4       | 32       |
| 1      | 2020-11-28 | 55      | 200      |
| 1      | 2020-12-03 | 1       | 42       |
| 2      | 2020-11-28 | 3       | 33       |
| 2      | 2020-12-09 | 47      | 74       |

Output:

| day        | emp_id | total_time |
|------------|--------|------------|
| 2020-11-28 | 1      | 173        |
| 2020-11-28 | 2      | 30         |
| 2020-12-03 | 1      | 41         |
| 2020-12-09 | 2      | 27         |

## 💡 Key Concepts

1. **Time Duration Calculation**:
   - The time spent in the office for an entry is calculated by subtracting `in_time` from `out_time`.

2. **Grouping by Employee and Day**:
   - Use `groupby(['event_day', 'emp_id'])` to group data by day and employee.

3. **Aggregation**:
   - Use `.sum()` to get the total time spent per employee per day.

4. **Renaming Columns**:
   - Rename columns to match the required output format (`event_day` to `day` and `duration` to `total_time`).

```python
import pandas as pd

def total_time(employees: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Compute duration of each entry
    employees['duration'] = employees['out_time'] - employees['in_time']
    
    # Step 2: Group by employee and day, sum durations
    result = employees.groupby(['event_day', 'emp_id'], as_index=False)['duration'].sum()
    
    # Step 3: Rename columns to match expected output
    result.rename(columns={'event_day': 'day', 'duration': 'total_time'}, inplace=True)
    
    return result
```

## 🔎 Solution Explanation

1. **Time Calculation**:
   - We calculate the duration by subtracting `in_time` from `out_time`.

2. **Grouping**:
   - Group by `event_day` and `emp_id` to ensure that each employee's time is aggregated per day.

3. **Aggregation**:
   - We use `.sum()` to calculate the total time for each group (each employee per day).

4. **Renaming**:
   - The final output requires the column names `day` and `total_time`, so we rename them accordingly.

**Time Complexity:** O(n) due to grouping and aggregation  
**Space Complexity:** O(n)

```python
# Sample Employees DataFrame
data = {
    "emp_id": [1, 1, 1, 2, 2],
    "event_day": ['2020-11-28', '2020-11-28', '2020-12-03', '2020-11-28', '2020-12-09'],
    "in_time": [4, 55, 1, 3, 47],
    "out_time": [32, 200, 42, 33, 74]
}

employees = pd.DataFrame(data)

# Running the function
total_time(employees)
```
