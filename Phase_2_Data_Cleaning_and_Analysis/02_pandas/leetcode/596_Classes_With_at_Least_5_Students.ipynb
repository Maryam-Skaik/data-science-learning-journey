# 596. Classes With at Least 5 Students

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – GroupBy / Filtering  

🔗 Problem Link: https://leetcode.com/problems/classes-with-at-least-5-students/

## 📝 Problem Description

Table: `Courses`

Columns:
- student
- class

Each row represents a student enrolled in a specific class.

**Task:**

Find all classes that have **at least 5 students enrolled**.

Return a table containing:
- class

## 💡 Key Concepts

1. **Grouping Data**
   - Use `groupby('class')` to group rows by class name.

2. **Counting Students**
   - Use `.count()` on the `student` column to count the number of students in each class.

3. **Filtering Results**
   - Keep only classes where the student count is **greater than or equal to 5**.

4. **Formatting Output**
   - Convert the result back into a DataFrame using `reset_index()`.

```python
import pandas as pd

def find_classes(courses: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Count students in each class
    counts = courses.groupby('class')['student'].count()
    
    # Step 2: Filter classes with at least 5 students
    result = counts[counts >= 5]
    
    # Step 3: Convert to DataFrame with correct format
    return result.reset_index()[['class']]
```

## 🔎 Solution Explanation

1. **Group by class**
   - `groupby('class')` groups all rows belonging to the same class.

2. **Count students**
   - `.count()` counts the number of student entries in each class.

3. **Filter classes**
   - `counts >= 5` keeps only classes that have at least five students.

4. **Return result**
   - `reset_index()` converts the Series result into a DataFrame.
   - Only the `class` column is returned to match the expected output format.

**Time Complexity:** O(n)  
**Space Complexity:** O(k) where *k* is the number of classes.

```python
# Example dataset
data = {
    "student": ["A","B","C","D","E","F","G","H","I"],
    "class": ["Math","English","Math","Biology","Math","Computer","Math","Math","Math"]
}

courses = pd.DataFrame(data)

# Run solution
find_classes(courses)
```
