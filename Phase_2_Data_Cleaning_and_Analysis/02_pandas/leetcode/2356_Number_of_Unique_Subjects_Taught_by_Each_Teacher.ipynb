## 📝 Problem Description

Table: `Teacher`

Columns:
- teacher_id
- subject_id
- dept_id

Each row indicates that a teacher teaches a subject in a specific department.

**Task:**

Calculate the number of **unique subjects** each teacher teaches in the university.

Even if the same subject is taught in multiple departments, it should only be counted **once per teacher**.

Return the result table containing:
- teacher_id
- cnt (number of unique subjects taught)

## 💡 Key Concepts

1. **Grouping Data**:
   - Use `groupby('teacher_id')` to group rows belonging to the same teacher.

2. **Counting Unique Values**:
   - Use `.nunique()` on `subject_id` to count distinct subjects.

3. **Formatting the Result**:
   - Convert the grouped result to a DataFrame using `reset_index()`.
   - Rename the column to match the required output (`cnt`).

```python
import pandas as pd

def count_unique_subjects(teacher: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Group by teacher_id and count unique subjects
    df = teacher.groupby('teacher_id')['subject_id'].nunique()
    
    # Step 2: Convert Series to DataFrame
    df = df.reset_index()
    
    # Step 3: Rename column to match expected output
    df = df.rename(columns={'subject_id': 'cnt'})
    
    return df
```

## 🔎 Solution Explanation

1. **Grouping by teacher**:
   - `groupby('teacher_id')` groups all rows belonging to the same teacher.

2. **Counting unique subjects**:
   - `.nunique()` counts the number of distinct `subject_id` values for each teacher.
   - This ensures that the same subject taught in different departments is counted only once.

3. **Formatting the result**:
   - `reset_index()` converts the grouped result into a regular DataFrame.
   - The column name is renamed to `cnt` to match the required output format.

**Time Complexity:** O(n)  
**Space Complexity:** O(k) where k is the number of teachers

```python
# Sample Teacher DataFrame
data = {
    "teacher_id": [1, 1, 1, 2, 2, 2, 2],
    "subject_id": [2, 2, 3, 1, 2, 3, 4],
    "dept_id": [3, 4, 3, 1, 1, 1, 1]
}

teacher = pd.DataFrame(data)

# Running the function
count_unique_subjects(teacher)
```
