# 1484. Group Sold Products By The Date

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – GroupBy / Aggregation  

🔗 Problem Link: https://leetcode.com/problems/group-sold-products-by-the-date/

## 📝 Problem Description

Table: `Activities`

Columns:
- sell_date
- product

Each row represents a product sold on a specific date.  
The table **may contain duplicate rows**.

**Task:**

For each `sell_date`:

1. Count the number of **distinct products sold**
2. Return the **product names sorted lexicographically**
3. Join product names using **comma ( , )**

Return a table containing:

- sell_date
- num_sold
- products

The result should be **ordered by `sell_date`**.

## 💡 Key Concepts

1. **Grouping Data**
   - Use `groupby('sell_date')` to group products sold on the same date.

2. **Counting Unique Products**
   - Use `.nunique()` to count the number of **distinct products**.

3. **Removing Duplicates**
   - Use `.unique()` to remove duplicate product names.

4. **Sorting Products**
   - Use `sorted()` to arrange product names **lexicographically**.

5. **Formatting Output**
   - Use `','.join()` to combine product names into a single string.

```python
import pandas as pd

def categorize_products(activities: pd.DataFrame) -> pd.DataFrame:
    result = (
        activities
        .groupby('sell_date')['product']
        .agg([
            ('num_sold', lambda x: x.nunique()),
            ('products', lambda x: ','.join(sorted(x.unique())))
        ])
        .reset_index()
        .sort_values('sell_date')
    )
    
    return result
```

## 🔎 Solution Explanation

1. **Group data by sell date**
   - `groupby('sell_date')` groups all products sold on the same date.

2. **Count unique products**
   - `.nunique()` calculates how many **distinct products** were sold that day.

3. **Remove duplicates**
   - `.unique()` ensures each product name appears only once.

4. **Sort product names**
   - `sorted()` arranges product names alphabetically.

5. **Create the product list**
   - `','.join()` combines the sorted product names into one comma-separated string.

6. **Format the result**
   - `.reset_index()` converts the grouped index back into a column.
   - `.sort_values('sell_date')` ensures the output is ordered by date.

**Time Complexity:** O(n log k) due to sorting products per date  
**Space Complexity:** O(k) where *k* is the number of unique products per date.

```python
# Example dataset
data = {
    "sell_date": [
        "2020-05-30", "2020-06-01", "2020-06-02",
        "2020-05-30", "2020-06-01", "2020-06-02",
        "2020-05-30"
    ],
    "product": [
        "Headphone", "Pencil", "Mask",
        "Basketball", "Bible", "Mask",
        "T-Shirt"
    ]
}

activities = pd.DataFrame(data)

# Run solution
categorize_products(activities)
```
