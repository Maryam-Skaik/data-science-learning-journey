## 📝 Problem Description

We are given a Pandas DataFrame `patients` with columns:

- patient_id
- patient_name
- conditions

The `conditions` column contains 0 or more condition codes separated by spaces.

We must return patients who have **Type I Diabetes**.

Important rule:
- Type I Diabetes condition codes always start with the prefix `DIAB1`.

Return:
- patient_id
- patient_name
- conditions

Result can be in any order.

## 💡 Key Concepts

1. Conditions are space-separated codes.
2. We must detect codes that:
   - Start with `DIAB1`
   - Are standalone codes (not partial matches inside other words).
3. Use Regular Expressions with `.str.contains()`.

```python
import pandas as pd

def find_patients(patients: pd.DataFrame) -> pd.DataFrame:
    pattern = r'(^|\s)DIAB1\d*(?=\s|$)'
    result = patients[patients['conditions'].str.contains(pattern, regex=True, na=False)]
    return result
```

## 🔎 Regex Pattern Explanation

Pattern used:

(^|\s)DIAB1\d*(?=\s|$)

Breakdown:

1. (^|\s)
   - ^  → Start of the string
   - \s → OR a whitespace character
   This ensures the code starts either at the beginning 
   or right after a space (separate word).

2. DIAB1
   - Matches the required prefix exactly.

3. \d*
   - \d  → any digit (0–9)
   - *   → zero or more digits
   This allows codes like:
     DIAB1
     DIAB100
     DIAB12345

4. (?=\s|$)
   - Positive lookahead
   - Ensures the match is followed by:
       • a space
       • OR end of string
   This prevents partial matches inside longer words.

Why this works:
- Matches full condition codes only.
- Avoids matching things like: "XDIAB100" or "DIAB100X".

## 🔎 Solution Explanation

Step 1:
Create a regex pattern to match a full condition code starting with DIAB1.

Step 2:
Use `.str.contains(pattern, regex=True, na=False)` to filter matching rows.

Step 3:
Return filtered DataFrame.

Time Complexity: O(n) <br>
Space Complexity: O(n)

```python
data = {
    "patient_id": [1,2,3,4,5],
    "patient_name": ["Daniel","Alice","Bob","George","Alain"],
    "conditions": ["YFEV COUGH", "", "DIAB100 MYOP", "ACNE DIAB100", "DIAB201"]
}

patients = pd.DataFrame(data)

find_patients(patients)
```

This problem reinforces:

- Word-boundary matching using regex
- Lookahead assertions (?=...)
- Filtering DataFrames with `.str.contains()`
- Handling NaN safely using `na=False`
