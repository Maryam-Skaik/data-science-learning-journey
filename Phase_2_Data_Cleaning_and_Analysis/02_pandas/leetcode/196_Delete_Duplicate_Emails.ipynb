# 196. Delete Duplicate Emails

## 📝 Problem Description

Table: `Person`

| Column Name | Type    |
|-------------|---------|
| id          | int     |
| email       | varchar |

- `id` is the primary key.  
- Each row contains an email. Emails are all lowercase.

**Task:**  

Delete all duplicate emails, keeping **only one row with the smallest id**.  

- For SQL, use a `DELETE` statement.  
- For Pandas, modify the `Person` DataFrame in place.  
- After running the code, the final `Person` table is returned.  
- Order does not matter.

**Example Input:**

| id | email            |
|----|-----------------|
| 1  | john@example.com |
| 2  | bob@example.com  |
| 3  | john@example.com |

**Expected Output:**

| id | email            |
|----|-----------------|
| 1  | john@example.com |
| 2  | bob@example.com  |

**Explanation:** `john@example.com` appears twice; keep the row with the smallest id = 1.

## 💡 Key Concepts

1. **Sorting**:
   - Sort the DataFrame by `id` ascending so that the smallest id is kept first.

2. **Dropping duplicates**:
   - Use `drop_duplicates(subset=['email'], keep='first')` to remove duplicate emails.  
   - `inplace=True` modifies the original DataFrame directly.

3. **Result**:
   - The resulting `Person` DataFrame contains only unique emails with the smallest id.

```python
import pandas as pd

def delete_duplicate_emails(person: pd.DataFrame) -> None:
    # Step 1: Sort by id ascending
    person.sort_values(by='id', ascending=True, inplace=True)
    
    # Step 2: Drop duplicate emails, keep the first (smallest id)
    person.drop_duplicates(subset=['email'], keep='first', inplace=True)
```

## 🔎 Solution Explanation

1. **Sorting**:
   - Sorting by `id` ensures that the first occurrence of each email has the smallest id.

2. **Dropping duplicates**:
   - `drop_duplicates` removes all subsequent rows with the same email while keeping the first one.

3. **In-place modification**:
   - `inplace=True` updates the original `Person` DataFrame as required by the problem.

**Time Complexity:** O(n log n) due to sorting  
**Space Complexity:** O(1) (modifying in place)

```python
# Sample Person DataFrame
data = {
    'id': [1, 2, 3],
    'email': ['john@example.com', 'bob@example.com', 'john@example.com']
}

person = pd.DataFrame(data)

# Running the function
delete_duplicate_emails(person)
person
```
