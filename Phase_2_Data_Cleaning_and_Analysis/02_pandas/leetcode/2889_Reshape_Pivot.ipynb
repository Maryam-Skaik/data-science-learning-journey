## 📝 Problem Description

We are given a Pandas DataFrame called `weather` with columns:

- city
- month
- temperature

Our task is to pivot the data so that:

- Each row represents a month
- Each city is a separate column
- Values are the temperatures

## 💡 Key Concept

In Pandas, pivoting a DataFrame can be done using `pd.pivot()`:

```python
pivoted = pd.pivot(df, index='row_identifier', columns='column_identifier', values='values_column')
```

- `index` → Rows
- `columns` → Columns
- `values` → Values to fill the table



````python
import pandas as pd

def pivotTable(weather: pd.DataFrame) -> pd.DataFrame:
    table = pd.pivot(weather, index='month', columns='city', values='temperature')
    return table
````

## 🔎 Solution Explanation

Step 1: Use `pd.pivot()`:

```python
table = pd.pivot(weather, index='month', columns='city', values='temperature')
```

- `index='month'` → Each row represents a month
- `columns='city'` → Each city becomes a separate column
- `values='temperature'` → Fill the table with temperature values

Step 2: Return the pivoted DataFrame.

Time Complexity: O(n)

- Each row is read once to place it in the pivoted table

Space Complexity: O(n)

- A new DataFrame is created with rows as months and columns as cities

````python
data = {
    "city": ["Jacksonville"]*5 + ["ElPaso"]*5,
    "month": ["January", "February", "March", "April", "May"]*2,
    "temperature": [13, 23, 38, 5, 34, 20, 6, 26, 2, 43]
}

weather = pd.DataFrame(data)

pivotTable(weather)
````
