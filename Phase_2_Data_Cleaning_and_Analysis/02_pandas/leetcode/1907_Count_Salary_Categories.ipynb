## 📝 Problem Description

Table: `Accounts`

Columns:
- account_id (primary key)
- income

Each row contains the monthly income of one bank account.

**Task:**

Calculate the number of bank accounts for each salary category:

- "Low Salary": income strictly less than 20000
- "Average Salary": income between 20000 and 50000 (inclusive)
- "High Salary": income strictly greater than 50000

The result must contain all three categories.
If a category has no accounts, return 0.

Return the result in any order.

## 💡 Key Concepts

1. **Boolean Conditions**:
   - Use comparison operators to define salary ranges.

2. **Counting with Conditions**:
   - Boolean expressions return True/False values.
   - Using `.sum()` counts the number of True values.

3. **Building Result DataFrame**:
   - Create a new DataFrame containing all three salary categories explicitly.

## 💡 Key Concepts

1. **Boolean Conditions**:
   - Use comparison operators to define salary ranges.

2. **Counting with Conditions**:
   - Boolean expressions return True/False values.
   - Using `.sum()` counts the number of True values.

3. **Building Result DataFrame**:
   - Create a new DataFrame containing all three salary categories explicitly.

```python
import pandas as pd

def count_salary_categories(accounts: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Count accounts in each salary range
    low = (accounts['income'] < 20000).sum()
    average = ((accounts['income'] >= 20000) & 
               (accounts['income'] <= 50000)).sum()
    high = (accounts['income'] > 50000).sum()

    # Step 2: Create result DataFrame
    return pd.DataFrame({
        'category': ['Low Salary', 'Average Salary', 'High Salary'],
        'accounts_count': [low, average, high]
    })
```

## 🔎 Solution Explanation

1. **Low Salary**:
   - Count incomes strictly less than 20000.

2. **Average Salary**:
   - Count incomes between 20000 and 50000 inclusive.

3. **High Salary**:
   - Count incomes strictly greater than 50000.

4. **Ensuring all categories appear**:
   - We manually construct the result DataFrame with all three categories,
     so even if a category has zero accounts, it appears with count 0.

**Time Complexity:** O(n)  
**Space Complexity:** O(1)

```python
# Sample Accounts DataFrame
data = {
    "account_id": [3, 2, 8, 6],
    "income": [108939, 12747, 87709, 91796]
}

accounts = pd.DataFrame(data)

# Running the function
count_salary_categories(accounts)
```
