## 📝 Problem Description

We are given a Pandas DataFrame called `products`.

Columns:
- name
- quantity
- price

Some rows have missing values in the `quantity` column.

Our task is to:
- Fill missing values in the `quantity` column with 0
- Return the updated DataFrame

## 💡 Key Concept

In Pandas, missing data can be filled using:

```python
df.fillna(value, inplace=True)
```

To fill missing values for a specific column:

```python
df.fillna({'column_name': value}, inplace=True)
```

- `inplace=True` modifies the original DataFrame directly
- Missing values are replaced with the specified value



````python
import pandas as pd

def fillMissingValues(products: pd.DataFrame) -> pd.DataFrame:
    products.fillna({'quantity': 0}, inplace=True)
    return products
````

## 🔎 Solution Explanation

Step 1: Use `fillna()` to replace missing values in a specific column:

```python
products.fillna({'quantity': 0}, inplace=True)
```

- `{'quantity': 0}` → Only the quantity column is affected
- Missing values are replaced with 0
- `inplace=True` → Changes the DataFrame directly

Step 2: Return the updated DataFrame.

Time Complexity: O(n), Pandas scans the `quantity` column once to fill missing values.

Space Complexity: O(1), No additional DataFrame is created because we use `inplace=True`.


````python
data = {
    "name": ["Wristwatch", "WirelessEarbuds", "GolfClubs", "Printer"],
    "quantity": [None, None, 779, 849],
    "price": [135, 821, 9319, 3051]
}

products = pd.DataFrame(data)

fillMissingValues(products)
````
