## 📝 Problem Description

We are given a Pandas DataFrame `employees` with columns:

- employee_id
- name
- salary

Bonus Rules:

1. Employee ID must be odd.
2. Employee name must NOT start with 'M'.
3. If both conditions are satisfied → bonus = 100% of salary.
4. Otherwise → bonus = 0.

Return a DataFrame with:
- employee_id
- bonus

Ordered by employee_id.

## 💡 Key Concepts

1. Check odd IDs:

```python
employees['employee_id'] % 2 == 1
```

2. Check first character of string:

```python
employees['name'].str[0] != 'M'
```

3. Assign values conditionally using `.loc`.

````python
import pandas as pd

def calculate_special_bonus(employees: pd.DataFrame) -> pd.DataFrame:
    filt = (employees.employee_id % 2 == 1) & (employees.name.str[0] != 'M')
    
    employees['bonus'] = 0
    employees.loc[filt, 'bonus'] = employees.loc[filt, 'salary']
    
    return employees[['employee_id', 'bonus']] \
            .sort_values('employee_id') \
            .reset_index(drop=True)
````

````python
def calculate_special_bonus(employees: pd.DataFrame) -> pd.DataFrame:
    employees['bonus'] = employees['salary'].where(
        (employees.employee_id % 2 == 1) & (employees.name.str[0] != 'M'),
        0
    )
    
    return employees[['employee_id', 'bonus']] \
            .sort_values('employee_id') \
            .reset_index(drop=True)
````

## 🔎 Solution Explanation

Step 1: Create condition:
- employee_id is odd
- name does not start with 'M'

Step 2: Assign bonus:
- Initialize bonus = 0
- Replace with salary where condition is True

Step 3: Sort by employee_id ascending.

Time Complexity: O(n)  
Space Complexity: O(n)

````python
data = {
    "employee_id": [2, 3, 7, 8, 9],
    "name": ["Meir", "Michael", "Addilyn", "Juan", "Kannon"],
    "salary": [3000, 3800, 7400, 6100, 7700]
}

employees = pd.DataFrame(data)

calculate_special_bonus(employees)
````
