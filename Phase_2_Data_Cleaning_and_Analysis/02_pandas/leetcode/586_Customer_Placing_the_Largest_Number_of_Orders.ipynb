## 📝 Problem Description

Table: `Orders`

Columns:
- order_number
- customer_number

Each row represents an order placed by a customer.

**Task:**

Find the `customer_number` of the customer who placed the **largest number of orders**.

The problem guarantees that exactly **one customer** has more orders than all others.

Return a table containing:
- customer_number

## 💡 Key Concepts

1. **Grouping Data**
   - Use `groupby('customer_number')` to group orders by customer.

2. **Counting Orders**
   - Use `.count()` on `order_number` to determine how many orders each customer placed.

3. **Finding Maximum**
   - Use `.max()` to determine the highest order count.

4. **Filtering**
   - Select the customer whose order count equals the maximum.

```python
import pandas as pd

def largest_orders(orders: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Count orders per customer
    counts = orders.groupby('customer_number')['order_number'].count()
    
    # Step 2: Find the maximum order count
    max_count = counts.max()
    
    # Step 3: Filter the customer with the maximum orders
    return counts[counts == max_count].reset_index()[['customer_number']]
```

## 🔎 Solution Explanation

1. **Group orders by customer**
   - `groupby('customer_number')` groups all orders belonging to the same customer.

2. **Count orders**
   - `.count()` counts how many orders each customer has placed.

3. **Find the largest order count**
   - `.max()` determines the highest number of orders among all customers.

4. **Filter the result**
   - We keep only the customer whose order count equals the maximum.

**Time Complexity:** O(n)  
**Space Complexity:** O(k) where *k* is the number of unique customers.

```python
# Example dataset
data = {
    "order_number": [1, 2, 3, 4],
    "customer_number": [1, 2, 3, 3]
}

orders = pd.DataFrame(data)

# Run solution
largest_orders(orders)
```
