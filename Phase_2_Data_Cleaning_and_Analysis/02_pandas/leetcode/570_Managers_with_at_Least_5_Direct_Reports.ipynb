# 570. Managers with at Least 5 Direct Reports

**Platform:** LeetCode  
**Difficulty:** Medium  
**Topic:** Pandas – GroupBy / Filtering  

🔗 Problem Link: https://leetcode.com/problems/managers-with-at-least-5-direct-reports/

## 📝 Problem Description

Table: `Employee`

Columns:
- id
- name
- department
- managerId

Each row represents an employee and their manager.

- `id` → employee ID  
- `managerId` → ID of the employee's manager  
- If `managerId` is `null`, the employee does not have a manager.

**Task:**

Find managers who have **at least five direct reports**.

Return a table containing:

- name

## 💡 Key Concepts

1. **Grouping Data**
   - Use `groupby('managerId')` to group employees by their manager.

2. **Counting Direct Reports**
   - Use `.size()` to count how many employees report to each manager.

3. **Filtering Managers**
   - Keep only managers where the report count is **greater than or equal to 5**.

4. **Matching Manager IDs**
   - Use `.isin()` to find employees whose `id` matches the filtered manager IDs.

5. **Selecting Output**
   - Return only the `name` column.

```python
import pandas as pd

def find_managers(employee: pd.DataFrame) -> pd.DataFrame:
    # Count direct reports for each manager
    report_counts = employee.groupby("managerId").size().reset_index(name="count")
    
    # Managers with at least 5 reports
    managers = report_counts[report_counts["count"] >= 5]["managerId"]
    
    # Get their names
    result = employee[employee["id"].isin(managers)][["name"]]
    
    return result
```

## 🔎 Solution Explanation

1. **Count direct reports**

   - `groupby("managerId").size()` counts how many employees report to each manager.

2. **Filter managers**

   - Keep only rows where the count is **≥ 5**.

3. **Identify manager IDs**

   - Extract the `managerId` values that meet the condition.

4. **Retrieve manager names**

   - Match these IDs with the `id` column in the employee table.

5. **Return result**

   - Select only the `name` column.

**Time Complexity:** O(n)  
**Space Complexity:** O(m) where *m* is the number of managers.

```python
# Example dataset

data = {
    "id": [101,102,103,104,105,106],
    "name": ["John","Dan","James","Amy","Anne","Ron"],
    "department": ["A","A","A","A","A","B"],
    "managerId": [None,101,101,101,101,101]
}

employee = pd.DataFrame(data)

find_managers(employee)
```
