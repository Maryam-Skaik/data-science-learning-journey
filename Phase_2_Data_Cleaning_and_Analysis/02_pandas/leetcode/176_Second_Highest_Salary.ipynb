## 📝 Problem Description

Given a Pandas DataFrame `Employee` with columns:

- id
- salary

Write a solution to find the **second highest distinct salary**.  

Return `None` if no second highest exists.

Result column: `SecondHighestSalary`.

## 💡 Key Concepts

1. **Remove duplicates** to handle distinct salaries:

```python
employee['salary'].drop_duplicates()
```

2. **Sort descending** to get highest salaries first:

```python
.sort_values(ascending=False)
```

3. **Access the 2nd element** (index 1) if it exists.
4. Return `None` if there are fewer than 2 distinct salaries.



````python
import pandas as pd

def second_highest_salary(employee: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Extract distinct salaries in descending order
    salaries = employee['salary'].drop_duplicates().sort_values(ascending=False)
    
    # Step 2: Check if 2nd highest exists
    if len(salaries) < 2:
        return pd.DataFrame({'SecondHighestSalary': [None]})
    
    # Step 3: Return 2nd highest salary
    return pd.DataFrame({'SecondHighestSalary': [salaries.iloc[1]]})
````

## 🔎 Solution Explanation

1. Remove duplicate salaries using `.drop_duplicates()` to ensure distinct values.
2. Sort descending with `.sort_values(ascending=False)` so highest comes first.
3. Check if there is at least 2 salaries:
   - If not → return `None`.
   - Otherwise → select index 1 (second element).
4. Return as a DataFrame with column `SecondHighestSalary`.

**Time Complexity:** O(n log n) due to sorting  
**Space Complexity:** O(n)

````python
data = {
    "id": [1, 2, 3, 4],
    "salary": [100, 200, 300, 200]
}

employee = pd.DataFrame(data)

second_highest_salary(employee)
````
