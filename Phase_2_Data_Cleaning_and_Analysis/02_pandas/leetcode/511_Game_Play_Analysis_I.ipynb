## 📝 Problem Description

Table: `Activity`

Columns:
- player_id (player's ID)
- device_id (device used for the game)
- event_date (the date of the activity)
- games_played (the number of games played)

**Task:**

Find the first login date for each player. Each row is a record of a player logging in and playing a number of games on a specific date.  

Return the result table containing `player_id` and `first_login`.

**Example:**

Input:

| player_id | device_id | event_date | games_played |
|-----------|-----------|------------|--------------|
| 1         | 2         | 2016-03-01 | 5            |
| 1         | 2         | 2016-05-02 | 6            |
| 2         | 3         | 2017-06-25 | 1            |
| 3         | 1         | 2016-03-02 | 0            |
| 3         | 4         | 2018-07-03 | 5            |

Output:

| player_id | first_login |
|-----------|-------------|
| 1         | 2016-03-01  |
| 2         | 2017-06-25  |
| 3         | 2016-03-02  |

## 💡 Key Concepts

1. **Sorting**:
   - Sort the data by `event_date` to ensure that the earliest event comes first.

2. **Dropping Duplicates**:
   - Use `drop_duplicates()` to keep the first occurrence of each `player_id`, which will correspond to their first login.

3. **Extracting First Login**:
   - After sorting and dropping duplicates, assign the `event_date` as `first_login` for each player.

```python
import pandas as pd

def game_analysis(activity: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Sort by event_date to get the earliest login date for each player
    activity = activity.sort_values('event_date')
    
    # Step 2: Drop duplicates, keeping only the first login for each player
    activity = activity.drop_duplicates('player_id')
    
    # Step 3: Set the first_login date to event_date for each player
    activity['first_login'] = activity['event_date']
    
    # Step 4: Return only player_id and first_login columns
    return activity[['player_id', 'first_login']]
```

## 🔎 Solution Explanation

1. **Sorting**:
   - We sort the `Activity` DataFrame by `event_date`, so the first login for each player is at the top of their respective group.

2. **Dropping Duplicates**:
   - `drop_duplicates('player_id')` ensures that we keep only the earliest record for each player, which represents their first login.

3. **Assigning the first login date**:
   - After dropping duplicates, we simply set the `first_login` column to be equal to `event_date` for each player.

**Time Complexity:** O(n log n) due to sorting  
**Space Complexity:** O(n) for the DataFrame storage

```python
# Sample Activity DataFrame
data = {
    "player_id": [1, 1, 2, 3, 3],
    "device_id": [2, 2, 3, 1, 4],
    "event_date": ['2016-03-01', '2016-05-02', '2017-06-25', '2016-03-02', '2018-07-03'],
    "games_played": [5, 6, 1, 0, 5]
}

activity = pd.DataFrame(data)

# Running the function
game_analysis(activity)
```
