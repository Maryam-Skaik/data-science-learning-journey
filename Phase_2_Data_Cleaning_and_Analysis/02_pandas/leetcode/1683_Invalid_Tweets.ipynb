## 📝 Problem Description

We are given a Pandas DataFrame `tweets` with columns:

- tweet_id
- content

A tweet is considered invalid if the length of its content is strictly greater than 15 characters.

Task:
Return a DataFrame containing only the `tweet_id` of invalid tweets.

## 💡 Key Concept

In Pandas, we can compute string length using:

```python
df['column'].str.len()
```

Then filter using a condition:

```python
df[df['column'].str.len() > value]
```

````python
import pandas as pd

def invalid_tweets(tweets: pd.DataFrame) -> pd.DataFrame:
    filt = tweets['content'].str.len() > 15
    tweets = tweets[filt]
    return tweets[['tweet_id']]
````

````python
def invalid_tweets(tweets: pd.DataFrame) -> pd.DataFrame:
    return tweets[tweets['content'].str.len() > 15][['tweet_id']]
````

## 🔎 Solution Explanation

Step 1: Calculate the length of each tweet:

```python
tweets['content'].str.len()
```

Step 2: Filter tweets with length > 15:

```python
tweets[tweets['content'].str.len() > 15]
```

Step 3: Return only the `tweet_id` column.

Time Complexity: O(n)

- `n` = number of tweets

Space Complexity: O(n)

````python
data = {
    "tweet_id": [1, 2, 3],
    "content": ["Let us Code", "I am a data scientist, I have no life ):", "Maryam's sad story"]
}

tweets = pd.DataFrame(data)

invalid_tweets(tweets)
````
