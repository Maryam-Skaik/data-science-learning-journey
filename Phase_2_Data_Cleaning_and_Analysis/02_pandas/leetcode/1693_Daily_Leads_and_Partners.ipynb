## 📝 Problem Description

Table: `DailySales`

Columns:
- date_id
- make_name
- lead_id
- partner_id

Each row represents a lead and partner for a product sold on a specific date.  
The table may contain duplicate rows.

**Task:**

For each combination of `date_id` and `make_name`:

1. Count the number of **distinct lead_id** → `unique_leads`  
2. Count the number of **distinct partner_id** → `unique_partners`

Return a table containing:

- date_id
- make_name
- unique_leads
- unique_partners

## 💡 Key Concepts

1. **Grouping Data**
   - Use `groupby(['date_id', 'make_name'])` to group by date and product.

2. **Counting Unique Values**
   - Use `.nunique()` to count distinct `lead_id` and `partner_id`.

3. **Aggregation**
   - Use `.agg()` with named aggregations to compute both counts in one step.

4. **Reset Index**
   - Convert grouped index back to columns using `.reset_index()`.

```python
import pandas as pd

def daily_leads_and_partners(daily_sales: pd.DataFrame) -> pd.DataFrame:
    result = (
        daily_sales
        .groupby(['date_id', 'make_name'])
        .agg(
            unique_leads=('lead_id', 'nunique'),
            unique_partners=('partner_id', 'nunique')
        )
        .reset_index()
    )
    
    return result
```

## 🔎 Solution Explanation

1. **Group by date and make**
   - `groupby(['date_id', 'make_name'])` collects all rows for each product per date.

2. **Count unique leads and partners**
   - `.agg(unique_leads=('lead_id', 'nunique'), unique_partners=('partner_id', 'nunique'))` counts distinct IDs.

3. **Format result**
   - `.reset_index()` restores `date_id` and `make_name` as columns.

**Time Complexity:** O(n)  
**Space Complexity:** O(k) where *k* is the number of unique (date_id, make_name) combinations.

```python
# Example dataset
data = {
    "date_id": [
        "2020-12-8", "2020-12-8", "2020-12-8",
        "2020-12-7", "2020-12-7",
        "2020-12-8", "2020-12-8",
        "2020-12-7", "2020-12-7", "2020-12-7"
    ],
    "make_name": [
        "toyota", "toyota", "toyota",
        "toyota", "toyota",
        "honda", "honda",
        "honda", "honda", "honda"
    ],
    "lead_id": [0, 1, 1, 0, 0, 1, 2, 0, 1, 2],
    "partner_id": [1, 0, 2, 2, 1, 2, 1, 1, 2, 1]
}

daily_sales = pd.DataFrame(data)

# Run solution
daily_leads_and_partners(daily_sales)
```
