## 📝 Problem Description

Given two Pandas DataFrames: 

1. `Employee` table:
   - `id`: employee ID
   - `name`: employee name
   - `salary`: employee salary
   - `departmentId`: department ID

2. `Department` table:
   - `id`: department ID
   - `name`: department name

Write a solution to find **employees who have the highest salary** in each department.

Return the result table with the following columns:

- Department
- Employee
- Salary

## 💡 Key Concepts

1. **Merging**:
   - Combine `Employee` and `Department` using `merge()` on the `departmentId`.

2. **Grouping**:
   - Use `groupby()` to find the highest salary per department.
   - Use `transform()` to broadcast the max salary per group.

3. **Filtering**:
   - Filter employees whose salary matches the maximum salary in their department.

```python
# Merge employee data with department
df = employee.merge(department, left_on='departmentId', right_on='id')

# Find max salary per department
max_salary = df.groupby('departmentId')['salary'].transform('max')

# Filter employees with max salary
df[df['salary'] == max_salary]
```

````python
import pandas as pd

def department_highest_salary(employee: pd.DataFrame, department: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Merge Employee and Department tables
    df = employee.merge(
        department,
        left_on='departmentId',
        right_on='id',
        suffixes=('_emp', '_dept')
    )
    
    # Step 2: Find max salary per department
    max_salary = df.groupby('departmentId')['salary'].transform('max')
    
    # Step 3: Filter employees with max salary in their department
    result = df[df['salary'] == max_salary]
    
    # Step 4: Format result
    return result[['name_dept', 'name_emp', 'salary']] \
        .rename(columns={
            'name_dept': 'Department',
            'name_emp': 'Employee',
            'salary': 'Salary'
        })
````

## 🔎 Solution Explanation

1. **Merging**:
   - We merged the `Employee` and `Department` DataFrames using the `departmentId` from `Employee` and `id` from `Department`. This allows us to associate each employee with their department.
   
2. **Grouping by department**:
   - We used `groupby('departmentId')` to group employees by their department, then used `.transform('max')` to compute the highest salary for each department. This step makes sure every employee row in the merged DataFrame gets the highest salary in their department.

3. **Filtering employees with the max salary**:
   - After obtaining the max salary for each department, we filtered the DataFrame to keep only those employees whose salary matches the max salary in their department.

4. **Result formatting**:
   - The final result is formatted by renaming columns for better readability (`Department`, `Employee`, and `Salary`).

**Time Complexity:** O(n log n) due to sorting for merging and grouping.  
**Space Complexity:** O(n)

````python
employee_data = {
    "id": [1, 2, 3, 4, 5],
    "name": ["Joe", "Jim", "Henry", "Sam", "Max"],
    "salary": [70000, 90000, 80000, 60000, 90000],
    "departmentId": [1, 1, 2, 2, 1]
}

department_data = {
    "id": [1, 2],
    "name": ["IT", "Sales"]
}

employee = pd.DataFrame(employee_data)
department = pd.DataFrame(department_data)

# Running the function
department_highest_salary(employee, department)
````
