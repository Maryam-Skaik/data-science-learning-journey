# 1050. Actors and Directors Who Cooperated At Least Three Times

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – GroupBy / Filtering  

🔗 Problem Link: https://leetcode.com/problems/actors-and-directors-who-cooperated-at-least-three-times/

## 📝 Problem Description

Table: `ActorDirector`

Columns:
- actor_id
- director_id
- timestamp

Each row represents a collaboration between an actor and a director at a specific timestamp.

**Task:**

Find all pairs `(actor_id, director_id)` where the actor has collaborated with the director **at least three times**.

Return a table with:
- actor_id
- director_id

## 💡 Key Concepts

1. **Grouping Data**
   - Use `groupby(['actor_id', 'director_id'])` to group collaborations by actor-director pair.

2. **Counting Collaborations**
   - Use `.count()` on `timestamp` to determine how many times they worked together.

3. **Filtering**
   - Keep only pairs where the count is greater than or equal to 3.

4. **Reset Index**
   - Convert the grouped index back to columns with `.reset_index()`.

```python
import pandas as pd

def actors_and_directors(actor_director: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Count collaborations per actor-director pair
    result = (
        actor_director
        .groupby(['actor_id', 'director_id'])
        .agg(count=('timestamp', 'count'))
        .reset_index()
    )

    # Step 2: Filter pairs with at least 3 collaborations
    result = result[result['count'] >= 3]

    # Step 3: Return required columns
    return result[['actor_id', 'director_id']]
```

## 🔎 Solution Explanation

1. **Group by actor and director**
   - `groupby(['actor_id', 'director_id'])` collects all collaborations for each pair.

2. **Count collaborations**
   - `.count()` counts the number of timestamps, which represents the number of collaborations.

3. **Filter for at least 3**
   - Keep only pairs with `count >= 3`.

4. **Format result**
   - Select only `actor_id` and `director_id` columns.

**Time Complexity:** O(n)  
**Space Complexity:** O(k) where *k* is the number of unique actor-director pairs.

```python
# Example dataset
data = {
    "actor_id": [1, 1, 1, 1, 1, 2, 2],
    "director_id": [1, 1, 1, 2, 2, 1, 1],
    "timestamp": [0, 1, 2, 3, 4, 5, 6]
}

actor_director = pd.DataFrame(data)

# Run solution
actors_and_directors(actor_director)
```
