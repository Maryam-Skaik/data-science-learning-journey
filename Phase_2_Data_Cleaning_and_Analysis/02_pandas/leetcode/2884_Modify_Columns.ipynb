## 📝 Problem Description

We are given a Pandas DataFrame called `employees`.

Columns:
- name
- salary

A company intends to give its employees a pay rise.

Our task is to:
- Modify the `salary` column by multiplying each salary by 2.
- Return the updated DataFrame.


## 💡 Key Concept

In Pandas, we can modify a column by directly assigning new values:

```python
df['column_name'] = expression
```

To multiply the `salary` column by 2:

```python 
df['salary'] *= 2
```

This will apply the multiplication element-wise to the entire column.

````python
import pandas as pd

def modifySalaryColumn(employees: pd.DataFrame) -> pd.DataFrame:
    employees['salary'] *= 2
    return employees
````

## 🔎 Solution Explanation

Step 1: Directly modify the `salary` column by multiplying it by 2:

```python
employees['salary'] *= 2
```

- This applies the multiplication to each value in the `salary column.

Step 2: Return the updated DataFrame.

Time Complexity: O(n), Each value in the `salary` column is multiplied by 2.

Space Complexity: O(1), No new DataFrame is created because we modify the existing one in-place.

````python
data = {
    "name": ["Jack", "Piper", "Mia", "Ulysses"],
    "salary": [19666, 74754, 62509, 54866]
}

employees = pd.DataFrame(data)

modifySalaryColumn(employees)
````
