## 📝 Problem Description

We are given a Pandas DataFrame `users` with columns:

- user_id
- name

The names are inconsistently capitalized (some letters uppercase, some lowercase).  

**Goal:** Convert names so that only the first character is uppercase, and the rest are lowercase.  

Return the DataFrame ordered by `user_id`.

## 💡 Key Concepts

1. Convert string to lowercase:
   
```python
users['name'].str.lower()
```

2. Capitalize the first letter:

```python
users['name'].str.capitalize()
```

3. Sort by column:

```python
users.sort_values('user_id')
```

4. Reset the index for a clean DataFrame.

````python
import pandas as pd

def fix_names(users: pd.DataFrame) -> pd.DataFrame:
    users['name'] = users['name'].str.lower().str.capitalize()
    return users.sort_values('user_id').reset_index(drop=True)
````

### 🔎 Explanation

1. `.str.lower()` → converts all letters to lowercase  
2. `.str.capitalize()` → capitalizes the first letter  
3. `.sort_values('user_id')` → ensures output is ordered by user_id  
4. `.reset_index(drop=True)` → resets index after sorting  

✅ This approach is vectorized and efficient for large DataFrames.


````python
data = {
    "user_id": [1, 2, 3],
    "name": ["aLice", "bOB", "cHaRLie"]
}

users = pd.DataFrame(data)
fix_names(users)
````
