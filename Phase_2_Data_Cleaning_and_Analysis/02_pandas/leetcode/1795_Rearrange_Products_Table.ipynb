# 1795. Rearrange Products Table

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – Reshaping / Data Transformation  

🔗 Problem Link: https://leetcode.com/problems/rearrange-products-table/

## 📝 Problem Description

Table: `Products`

Columns:
- product_id (primary key)
- store1
- store2
- store3

Each row indicates the product's price in three different stores.
If a product is not available in a store, the price is null.

**Task:**

Rearrange the table so that each row contains:
- product_id
- store
- price

If a product is not available in a store, do not include that row.

Return the result in any order.

Example:

Input:
product_id = 0, store1 = 95, store2 = 100, store3 = 105  
product_id = 1, store1 = 70, store2 = null, store3 = 80  

Output:
(0, store1, 95)  
(0, store2, 100)  
(0, store3, 105)  
(1, store1, 70)  
(1, store3, 80)

## 💡 Key Concepts

1. **Reshaping (Unpivoting)**:
   - Use `melt()` to convert wide format (store1, store2, store3 columns) into long format.

2. **Handling Missing Values**:
   - Use `dropna()` to remove rows where the price is null.

3. **Result**:
   - The final DataFrame contains product_id, store, and price.

```python
import pandas as pd

def rearrange_products_table(products: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Unpivot store columns into rows
    df = products.melt(
        id_vars='product_id',
        value_vars=['store1', 'store2', 'store3'],
        var_name='store',
        value_name='price'
    )

    # Step 2: Remove rows where price is null
    df.dropna(inplace=True)

    return df
```

## 🔎 Solution Explanation

1. **Unpivoting the table**:
   - `melt()` transforms store1, store2, and store3 columns into a single column called `store`, with corresponding values in `price`.

2. **Removing unavailable products**:
   - `dropna()` removes rows where price is null, meaning the product is not available in that store.

3. **Final result**:
   - The output contains one row per available product-store combination.

**Time Complexity:** O(n)  
**Space Complexity:** O(n)

```python
# Sample Products DataFrame
data = {
    "product_id": [0, 1],
    "store1": [95, 70],
    "store2": [100, None],
    "store3": [105, 80]
}

products = pd.DataFrame(data)

# Running the function
rearrange_products_table(products)
```
