## 📝 Problem Description

Table: `Students`

Columns:
- student_id
- student_name

Each row represents a student in the school.

Table: `Subjects`

Columns:
- subject_name

Each row represents a subject offered in the school.

Table: `Examinations`

Columns:
- student_id
- subject_name

Each row indicates that a student attended an exam for a specific subject.
The table may contain duplicates.

**Task:**

Find how many times each student attended each subject's exam.

Important requirements:

- The result must include **all combinations of students and subjects**
- If a student did not attend an exam, return **0**
- Order the result by:
  - `student_id`
  - `subject_name`

Return a table containing:

- student_id
- student_name
- subject_name
- attended_exams

## 💡 Key Concepts

1. **Cross Join**
   - Use `merge(..., how='cross')` to create every possible student–subject pair.

2. **Counting Exam Attendance**
   - Add a column `attended_exams = 1` to represent each exam occurrence.

3. **Left Join**
   - Merge the student-subject pairs with the examinations table to include all combinations.

4. **Aggregation**
   - Use `groupby()` and `sum()` to count how many times each student attended each exam.

5. **Handling Missing Values**
   - Replace missing counts with `0`.

6. **Sorting**
   - Sort the final result by `student_id` and `subject_name`.

```python
import pandas as pd

def students_and_examinations(students: pd.DataFrame, subjects: pd.DataFrame, examinations: pd.DataFrame) -> pd.DataFrame:
    # create all student-subject combinations
    student_subject = students.merge(subjects, how='cross')
    
    # add counter for each exam record
    examinations = examinations.copy()
    examinations['attended_exams'] = 1
    
    # merge combinations with exams
    merged = student_subject.merge(examinations, how='left', on=['student_id', 'subject_name'])
    
    # group and count exams
    result = (
        merged
        .groupby(['student_id', 'student_name', 'subject_name'], dropna=False)['attended_exams']
        .sum()
        .fillna(0)
        .astype(int)
        .reset_index()
    )
    
    return result.sort_values(['student_id', 'subject_name']).reset_index(drop=True)
```

## 🔎 Solution Explanation

1. **Create all student-subject pairs**

   - `students.merge(subjects, how='cross')` generates every possible combination of students and subjects.

2. **Mark each exam occurrence**

   - Add `attended_exams = 1` so that each row represents one exam attendance.

3. **Merge exam records**

   - A left join ensures that all student-subject combinations remain, even if no exam exists.

4. **Count exam attendance**

   - `groupby(...).sum()` counts how many times each student attended each subject exam.

5. **Handle missing values**

   - `.fillna(0)` converts missing counts to `0`.

6. **Sort the result**

   - The final table is sorted by `student_id` and `subject_name`.

**Time Complexity:** O(S × C + E)  
- S = number of students  
- C = number of subjects  
- E = number of examination records  

**Space Complexity:** O(S × C)

```python
# Example datasets

students_data = {
    "student_id": [1, 2, 13, 6],
    "student_name": ["Alice", "Bob", "John", "Alex"]
}

subjects_data = {
    "subject_name": ["Math", "Physics", "Programming"]
}

examinations_data = {
    "student_id": [1,1,1,2,1,1,13,13,13,2,1],
    "subject_name": [
        "Math","Physics","Programming",
        "Programming","Physics","Math",
        "Math","Programming","Physics",
        "Math","Math"
    ]
}

students = pd.DataFrame(students_data)
subjects = pd.DataFrame(subjects_data)
examinations = pd.DataFrame(examinations_data)

students_and_examinations(students, subjects, examinations)
```
