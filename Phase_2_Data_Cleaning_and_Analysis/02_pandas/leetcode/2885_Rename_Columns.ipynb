## 📝 Problem Description

We are given a Pandas DataFrame called `students`.

Columns:
- id
- first
- last
- age

Our task is to rename the columns as follows:
- `id` → `student_id`
- `first` → `first_name`
- `last` → `last_name`
- `age` → `age_in_years`

The function should return the DataFrame with renamed columns.


## 💡 Key Concept

In Pandas, you can rename columns using the `rename()` method:

```python
df.rename(columns={'old_name': 'new_name'})
```

This method returns a new DataFrame with the renamed columns. You can apply it as follows:

```python
df.rename(columns={'id': 'student_id'})
```

To modify the DataFrame in place, use `inplace=True`:

```python
df.rename(columns={'id': 'student_id'}, inplace=True)
```

````python
import pandas as pd

def renameColumns(students: pd.DataFrame) -> pd.DataFrame:
    students = students.rename(columns={
        'id': 'student_id', 
        'first': 'first_name', 
        'last': 'last_name', 
        'age': 'age_in_years'
    })
    return students
````

## 🔎 Solution Explanation

Step 1: We use `rename()` to rename columns:

```python
students.rename(columns={'id': 'student_id', 'first': 'first_name', 'last': 'last_name', 'age': 'age_in_years'})
```

- The `columns` argument is a dictionary where:
    - The keys are the old column names
    - The values are the new column names

Step 2: The function returns the updated DataFrame with renamed columns.

Time Complexity: O(n), Each column name is checked and replaced once.

Space Complexity: O(1), No additional DataFrame is created since we return the modified one.

````python
data = {
    "id": [1, 2, 3, 4, 5],
    "first": ["Mason", "Ava", "Taylor", "Georgia", "Thomas"],
    "last": ["King", "Wright", "Hall", "Thompson", "Moore"],
    "age": [6, 7, 16, 18, 10]
}

students = pd.DataFrame(data)

renameColumns(students)
````
