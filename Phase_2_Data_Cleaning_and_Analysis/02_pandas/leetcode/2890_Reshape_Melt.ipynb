## 📝 Problem Description

We are given a Pandas DataFrame called `report` with columns:

- product
- quarter_1
- quarter_2
- quarter_3
- quarter_4

The data is in wide format.  

Our task is to reshape it into long format so that each row represents sales data for a product in a specific quarter:

- Columns in the result: `product`, `quarter`, `sales`

## 💡 Key Concept

In Pandas, reshaping from wide to long format can be done using `pd.melt()`:

```python
pd.melt(df, id_vars='id_column', var_name='new_column_name', value_name='value_column_name')
```

- `id_vars` → Column(s) to keep as identifier (not unpivoted)
- `var_name` → Name of the new column holding the original column names
- `value_name` → Name of the new column holding the values from the unpivoted columns



````python
import pandas as pd

def meltTable(report: pd.DataFrame) -> pd.DataFrame:
    table = pd.melt(report, id_vars='product', var_name='quarter', value_name='sales')
    return table
````

## 🔎 Solution Explanation

Step 1: Use `pd.melt()` to reshape the DataFrame:

```python
table = pd.melt(report, id_vars='product', var_name='quarter', value_name='sales')
```

- `id_vars='product'` → Keeps the product column as-is
- `var_name='quarter'` → Creates a new column with the original column names (quarter_1, quarter_2, etc.)
- `value_name='sales'` → Values from the original quarter columns are placed under sales

Step 2: Return the reshaped DataFrame.

Time Complexity: O(n*m)

- `n` = number of rows, `m` = number of quarters (columns being melted)

Space Complexity: O(n*m)

- A new long-format DataFrame is created


````python
data = {
    "product": ["Umbrella", "SleepingBag"],
    "quarter_1": [417, 800],
    "quarter_2": [224, 936],
    "quarter_3": [379, 93],
    "quarter_4": [611, 875]
}

report = pd.DataFrame(data)

meltTable(report)
````
