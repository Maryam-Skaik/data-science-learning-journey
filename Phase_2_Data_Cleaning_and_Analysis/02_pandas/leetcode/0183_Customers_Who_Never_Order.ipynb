## 📝 Problem Description

We are given two DataFrames:

### Customers
- id
- name

### Orders
- id
- customerId (foreign key referencing Customers.id)

Our task:  
Find all customers who never placed any order.

Return a DataFrame with one column:
- Customers

## 💡 Key Concept

To find unmatched records between two tables:

1. Perform a LEFT JOIN.
2. Check for NULL values in the joined column.
3. Select only the required column.

In Pandas:

```python
pd.merge(df1, df2, how='left', left_on='col1', right_on='col2')
```

Then use:

```python
.isna()
```


````python
import pandas as pd

def find_customers(customers: pd.DataFrame, orders: pd.DataFrame) -> pd.DataFrame:
    merged_df = pd.merge(customers, orders, how='left', left_on='id', right_on='customerId')
    
    result = merged_df[merged_df['customerId'].isna()]
    
    return pd.DataFrame({'Customers': result['name']})
````

## 🔎 Solution Explanation

Step 1: Perform LEFT JOIN:

```python
pd.merge(customers, orders, how='left', left_on='id', right_on='customerId')
```

- Keeps all customers
- Adds order data if it exists
- If no order → `customerId` becomes NaN

Step 2: Filter customers with no orders:

```python
merged_df['customerId'].isna()
```

Step 3: Return only the name column renamed to Customers.

Time Complexity: O(n + m)

- `n` = number of customers
- `m` = number of orders

Space Complexity: O(n)

````python
customers_data = {
    "id": [1, 2, 3, 4],
    "name": ["Joe", "Henry", "Sam", "Max"]
}

orders_data = {
    "id": [1, 2],
    "customerId": [3, 1]
}

customers = pd.DataFrame(customers_data)
orders = pd.DataFrame(orders_data)

find_customers(customers, orders)
````
