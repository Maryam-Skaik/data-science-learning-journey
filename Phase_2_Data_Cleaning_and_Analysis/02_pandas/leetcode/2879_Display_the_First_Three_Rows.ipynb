## 📝 Problem Description

We are given a Pandas DataFrame called `employees`.

The DataFrame contains columns such as:
- employee_id
- name
- department
- salary

Our task is to return a new DataFrame that contains only the first 3 rows.

The structure and column order must remain unchanged.


## 💡 Key Concept

In Pandas, to view the first rows of a DataFrame, we use:

```python
df.head(n)
```

- `n` → number of rows to display
- Default value is 5 if not specified

Example:

```python
df.head(3)
```

This returns the first 3 rows while preserving the DataFrame structure.

````python
import pandas as pd

def selectFirstRows(employees: pd.DataFrame) -> pd.DataFrame:
    return employees.head(3)
````

## 🔎 Solution Explanation

Step 1: The function receives a Pandas DataFrame called `employees`.

Step 2: We use:

```python
employees.head(3)
```

This returns:

- The first 3 rows
- All columns
- Original structure preserved

No modification is made to the original DataFrame.

Time Complexity: O(1), Accessing the first few rows is constant time.

Space Complexity: O(1)

````python
data = {
    "employee_id": [3, 90, 9, 60, 49, 43],
    "name": ["Bob", "Alice", "Tatiana", "Annabelle", "Jonathan", "Khaled"],
    "department": [
        "Operations",
        "Sales",
        "Engineering",
        "InformationTechnology",
        "HumanResources",
        "Administration"
    ],
    "salary": [48675, 11096, 33805, 37678, 23793, 40454]
}

employees = pd.DataFrame(data)

selectFirstRows(employees)
````
