## 📝 Problem Description

We are given a Pandas DataFrame called `customers`.

Columns:
- customer_id
- name
- email

Some rows contain duplicate emails.

Our task is to:
- Remove duplicate rows based on the `email` column
- Keep only the first occurrence
- Return the cleaned DataFrame


## 💡 Key Concept

In Pandas, duplicate rows can be removed using:

```python
df.drop_duplicates()
```

To remove duplicates based on a specific column:

```python
df.drop_duplicates(subset=['column_name'])
```

By default:

- keep='first'
- The first occurrence is retained

````python
import pandas as pd

def dropDuplicateEmails(customers: pd.DataFrame) -> pd.DataFrame:
    customers.drop_duplicates(subset=['email'], inplace=True)
    return customers
````

## 🔎 Solution Explanation

Step 1: Use `drop_duplicates()`.

```python
customers.drop_duplicates(subset=['email'], inplace=True)
```

- `subset=['email']` → Checks duplicates only in the email column.
- `inplace=True` → Modifies the original DataFrame directly.
- Default behavior keeps the first occurrence.

Step 2: Return the updated DataFrame.

Time Complexity: O(n), Pandas scans the email column once.

Space Complexity: O(1), No additional DataFrame is created because we use inplace=True.

````python
data = {
    "customer_id": [1, 2, 3, 4, 5, 6],
    "name": ["Ella", "David", "Zachary", "Alice", "Finn", "Violet"],
    "email": [
        "emily@example.com",
        "michael@example.com",
        "sarah@example.com",
        "john@example.com",
        "john@example.com",
        "alice@example.com"
    ]
}

customers = pd.DataFrame(data)

dropDuplicateEmails(customers)
````
