## 📝 Problem Description

We are given a Pandas DataFrame `products` with columns:

- product_id
- low_fats (`Y` or `N`)
- recyclable (`Y` or `N`)

Our task: find all products that are **both low fat and recyclable**.

Return only the `product_id` column. The order does not matter.

## 💡 Key Concept

In Pandas, we can filter rows using multiple conditions:

- AND condition: `&`
- OR condition: `|`

Example:

```python
df[(df['col1'] == 'value1') & (df['col2'] == 'value2')]
```

Select only needed columns:

```python
df[['column_name']]
```

````python
import pandas as pd

def find_products(products: pd.DataFrame) -> pd.DataFrame:
    filt = (products['low_fats'] == 'Y') & (products['recyclable'] == 'Y')
    products = products[filt]
    return products[['product_id']]
````

## 🔎 Solution Explanation

Step 1: Create a filter for low fat and recyclable products:

```python
filt = (products['low_fats'] == 'Y') & (products['recyclable'] == 'Y')
```

Step 2: Apply the filter:

```python
products = products[filt]
```

Step 3: Return only the `product_id` column:

```python
[['product_id']]
```

Time Complexity: O(n)

- `n` = number of products

Space Complexity: O(n)

- Filtered DataFrame is created

````python
data = {
    "product_id": [0, 1, 2, 3, 4],
    "low_fats": ['Y', 'Y', 'N', 'Y', 'N'],
    "recyclable": ['N', 'Y', 'Y', 'Y', 'N']
}

products = pd.DataFrame(data)

find_products(products)
````
