## 📝 Problem Description

We are given a Pandas DataFrame `world` with columns:

- name
- continent
- area
- population
- gdp

A country is considered **big** if:

1. Its area is at least 3,000,000 km², **or**
2. Its population is at least 25,000,000

We need to return a DataFrame with columns:

- name
- population
- area  

for all big countries. The result order does not matter.

## 💡 Key Concept

In Pandas, we can filter rows using boolean conditions:

```python
df[(df['column1'] > value1) | (df['column2'] > value2)]
```

- Use `|` for OR conditions
- Use `&` for AND conditions
- Select specific columns with `[['col1', 'col2']]`



````python
import pandas as pd

def big_countries(world: pd.DataFrame) -> pd.DataFrame:
    filt = (world['area'] >= 3000000) | (world['population'] >= 25000000)
    world = world[filt]
    return world[['name', 'population', 'area']]
````

## 🔎 Solution Explanation

Step 1: Apply the filter for big countries:

```python
filt = (world['area'] >= 3000000) | (world['population'] >= 25000000)
```

- Select countries with area ≥ 3,000,000 or population ≥ 25,000,000

Step 2: Filter the DataFrame:

```python
world = world[filt]
```

Step 3: Return only required columns:

```python
[['name', 'population', 'area']]
```

Time Complexity: O(n)

- `n` = number of countries (rows)

Space Complexity: O(n)

- Filtered DataFrame is created

````python
data = {
    "name": ["Afghanistan", "Albania", "Algeria", "Andorra", "Angola"],
    "continent": ["Asia", "Europe", "Africa", "Europe", "Africa"],
    "area": [652230, 28748, 2381741, 468, 1246700],
    "population": [25500100, 2831741, 37100000, 78115, 20609294],
    "gdp": [20343000000, 12960000000, 188681000000, 3712000000, 100990000000]
}

world = pd.DataFrame(data)

big_countries(world)
````
