## 📝 Problem Description

We are given a Pandas DataFrame called `students`.

Columns:
- student_id
- name
- age

Our task is to:
- Find the student with student_id = 101
- Return only the columns: name and age

The result must be returned as a DataFrame.

## 💡 Key Concepts

1️⃣ Filtering rows in Pandas:

```python
df[df['column'] == value]
```

2️⃣ Selecting specific columns:

```python
df[['col1', 'col2']]
```

3️⃣ Using `.loc[]` for row and column selection:

```python
df.loc[row_condition, [columns]]
```

- First argument → row condition
- Second argument → list of columns

````python
import pandas as pd

def selectData(students: pd.DataFrame) -> pd.DataFrame:
    return students.loc[students['student_id'] == 101, ['name', 'age']]
````

## 🔎 Solution Explanation

Step 1: Apply a row filter

```python
students['student_id'] == 101
```

This creates a boolean mask that selects only rows where student_id equals 101.

Step 2: Use `.loc[]` to:

- Filter rows using the condition
- Select only specific columns: ['name', 'age']

Final result:

- One row
- Two columns
- Returned as a DataFrame

Time Complexity: O(n), We scan the student_id column once.

Space Complexity: O(1)

````python
data = {
    "student_id": [101, 53, 128, 3],
    "name": ["Ulysses", "William", "Henry", "Henry"],
    "age": [13, 10, 6, 11]
}

students = pd.DataFrame(data)

selectData(students)
````
