## 📝 Problem Description

We are given two Pandas DataFrames, `df1` and `df2`, with the same columns:

- student_id
- name
- age

Our task is to concatenate them vertically to create a single DataFrame with all rows from both DataFrames.

## 💡 Key Concept

In Pandas, we can concatenate DataFrames vertically using:

```python
pd.concat([df1, df2])
```

- By default, `axis=0`, which stacks rows.
- If the columns are identical, the rows are combined seamlessly.
- The index is preserved by default; `ignore_index=True` can reset it.

````python
import pandas as pd

def concatenateTables(df1: pd.DataFrame, df2: pd.DataFrame) -> pd.DataFrame:
    df = pd.concat([df1, df2])
    return df
````

## 🔎 Solution Explanation

Step 1: Use `pd.concat()` to stack the DataFrames vertically:

```python
df = pd.concat([df1, df2])
```

- `df1` rows come first, followed by `df2`.
- Columns are aligned by name.
- The combined DataFrame `df` contains all rows.

Step 2: Return the combined DataFrame.

Time Complexity: O(n + m)

- `n` = number of rows in `df1`
- `m` = number of rows in `df2`

Space Complexity: O(n + m)

- A new DataFrame is created to hold all rows.

````python
data1 = {
    "student_id": [1, 2, 3, 4],
    "name": ["Mason", "Ava", "Taylor", "Georgia"],
    "age": [8, 6, 15, 17]
}

data2 = {
    "student_id": [5, 6],
    "name": ["Leo", "Alex"],
    "age": [7, 7]
}

df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)

concatenateTables(df1, df2)
````
