# 607. Sales Person

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – Merge / Filtering

## 📝 Problem Description

Table: `SalesPerson`

Columns:
- sales_id
- name
- salary
- commission_rate
- hire_date

Each row represents a salesperson and their information.

Table: `Company`

Columns:
- com_id
- name
- city

Each row represents a company.

Table: `Orders`

Columns:
- order_id
- order_date
- com_id
- sales_id
- amount

Each row represents an order made by a company through a salesperson.

**Task:**

Find the names of all salespersons who **did not have any orders** related to the company **"RED"**.

Return a table containing:

- name

## 💡 Key Concepts

1. **Joining Tables**
   - Use `merge()` to connect `Orders` with `Company` in order to identify which company each order belongs to.

2. **Filtering Rows**
   - Select orders where the company name is `"RED"`.

3. **Finding Relevant Salespersons**
   - Extract the `sales_id` values associated with `"RED"` orders.

4. **Excluding Salespersons**
   - Use `.isin()` with negation (`~`) to remove those salespersons from the result.

5. **Selecting Final Output**
   - Return only the `name` column.

```python
import pandas as pd

def sales_person(sales_person: pd.DataFrame, company: pd.DataFrame, orders: pd.DataFrame) -> pd.DataFrame:
    # Merge orders with company to identify company names
    orders_with_company = orders.merge(company, on='com_id', how='left')
    
    # Get salespersons who sold to the company 'RED'
    sales_with_red = orders_with_company.loc[
        orders_with_company['name'] == 'RED', 'sales_id'
    ].unique()
    
    # Select salespersons who never sold to 'RED'
    result = sales_person.loc[
        ~sales_person['sales_id'].isin(sales_with_red),
        ['name']
    ]
    
    return result
```

## 🔎 Solution Explanation

1. **Combine Orders and Company tables**

   - `merge()` connects orders with their company names using `com_id`.

2. **Find orders for company "RED"**

   - Filter rows where `name == 'RED'`.

3. **Extract salesperson IDs**

   - `.unique()` retrieves the salespersons who handled those orders.

4. **Exclude those salespersons**

   - Use `~sales_person['sales_id'].isin(...)` to keep only salespersons who never sold to `"RED"`.

5. **Return names**

   - Select only the `name` column for the final result.

**Time Complexity:** O(n + m)  
**Space Complexity:** O(n)

```python
# Example datasets

salesperson_data = {
    "sales_id": [1,2,3,4,5],
    "name": ["John","Amy","Mark","Pam","Alex"],
    "salary": [100000,12000,65000,25000,5000],
    "commission_rate": [6,5,12,25,10],
    "hire_date": ["2006-04-01","2010-05-01","2008-12-25","2005-01-01","2007-02-03"]
}

company_data = {
    "com_id": [1,2,3,4],
    "name": ["RED","ORANGE","YELLOW","GREEN"],
    "city": ["Boston","New York","Boston","Austin"]
}

orders_data = {
    "order_id": [1,2,3,4],
    "order_date": ["2014-01-01","2014-02-01","2014-03-01","2014-04-01"],
    "com_id": [3,4,1,1],
    "sales_id": [4,5,1,4],
    "amount": [10000,5000,50000,25000]
}

sales_person_df = pd.DataFrame(salesperson_data)
company_df = pd.DataFrame(company_data)
orders_df = pd.DataFrame(orders_data)

sales_person(sales_person_df, company_df, orders_df)
```
