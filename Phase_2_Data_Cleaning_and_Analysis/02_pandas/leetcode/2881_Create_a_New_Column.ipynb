## 📝 Problem Description

We are given a Pandas DataFrame called `employees`.

Columns:
- name
- salary

A company plans to give employees a bonus.

Our task is to:
- Create a new column called `bonus`
- Set its value to double the salary

The function should return the updated DataFrame.


## 💡 Key Concept

In Pandas, we can create or modify a column by assigning a value:

```python
df['new_column'] = expression
```


Pandas performs element-wise operations automatically.

Example:

```python
df['bonus'] = df['salary'] * 2
```

This multiplies each salary value by 2 and stores it in a new column.

````python
import pandas as pd

def createBonusColumn(employees: pd.DataFrame) -> pd.DataFrame:
    employees['bonus'] = employees['salary'] * 2
    return employees
````

## 🔎 Solution Explanation

Step 1: Access the salary column:

```python
employees['salary']
```

Step 2: Multiply it by 2.

Pandas automatically applies the multiplication to every row (vectorized operation).

Step 3: Assign the result to a new column:

```python
employees['bonus'] = employees['salary'] * 2
```

Step 4: Return the updated DataFrame.

Time Complexity: O(n), Each row’s salary is processed once.

Space Complexity: O(n), A new column of size n is created.

````python
data = {
    "name": ["Piper", "Grace", "Georgia", "Willow", "Finn", "Thomas"],
    "salary": [4548, 28150, 1103, 6593, 74576, 24433]
}

employees = pd.DataFrame(data)

createBonusColumn(employees)
````
