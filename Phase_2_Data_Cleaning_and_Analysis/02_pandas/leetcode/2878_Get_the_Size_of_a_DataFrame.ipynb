# 2878. Get the Size of a DataFrame

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – DataFrame Shape

## 📝 Problem Description

We are given a Pandas DataFrame called `players`.

The DataFrame contains multiple columns such as:
- player_id
- name
- age
- position
- team
- ...

Our task is to return:

[number_of_rows, number_of_columns]

### Example

If the DataFrame contains:
- 10 rows
- 5 columns

The output should be:

[10, 5]

## 💡 Key Concept

In Pandas, every DataFrame has an attribute called:

```python
df.shape
```

`shape` returns a tuple:

(number_of_rows, number_of_columns)

Example:

```python
df.shape  # (10, 5)
```

To access:

- Rows → df.shape[0]
- Columns → df.shape[1]

````python
import pandas as pd
from typing import List

def getDataframeSize(players: pd.DataFrame) -> List[int]:
    return [players.shape[0], players.shape[1]]
````

## 🔎 Solution Explanation

Step 1: The function receives a Pandas DataFrame called `players`.

Step 2: We use:

```python
players.shape
```

This returns:
(rows, columns)

Step 3: We extract:

- players.shape[0] → number of rows
- players.shape[1] → number of columns

Step 4: Return the result as a list:

[number_of_rows, number_of_columns]

Time Complexity: O(1), Accessing `.shape` is constant time.

Space Complexity: O(1)

````python
data = {
    "player_id": [846, 749, 155],
    "name": ["Mason", "Riley", "Bob"],
    "age": [21, 30, 28],
    "position": ["Forward", "Winger", "Striker"],
    "team": ["RealMadrid", "Barcelona", "ManchesterUnited"]
}

players = pd.DataFrame(data)

getDataframeSize(players)
````
