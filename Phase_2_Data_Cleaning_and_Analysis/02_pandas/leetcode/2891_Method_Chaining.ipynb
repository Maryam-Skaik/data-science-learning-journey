## 📝 Problem Description

We are given a Pandas DataFrame `animals` with columns:

- name
- species
- age
- weight

Our task:

1. List the names of animals that weigh strictly more than 100 kg.
2. Sort them by weight in descending order.
3. Use method chaining to perform this in a single line of code.

## 💡 Key Concept

Method chaining in Pandas allows multiple operations to be applied sequentially on a DataFrame without creating intermediate variables.

Example:

```python
df[df['column'] > value].sort_values(by='column_name')[['column_to_return']]
```

- Filter → `.sort_values()` → Select columns
- Keeps code concise and readable

````python
import pandas as pd

def findHeavyAnimals(animals: pd.DataFrame) -> pd.DataFrame:
    return animals[animals['weight'] > 100] \
                  .sort_values(by='weight', ascending=False) \
                  [['name']]
````

## 🔎 Solution Explanation

Step 1: Filter animals weighing more than 100 kg:

```python
animals[animals['weight'] > 100]
```

Step 2: Sort the filtered DataFrame by `weight` descending:

```python
.sort_values(by='weight', ascending=False)
```

Step 3: Select only the `name` column:

```python
[['name']]
```

- Combined in one line → method chaining
- Returns a DataFrame with sorted names of heavy animals

Time Complexity: O(n log n)

- Sorting dominates complexity

Space Complexity: O(n)

- Filtered DataFrame is created during the chain

````python
data = {
    "name": ["Tatiana", "Khaled", "Alex", "Jonathan", "Stefan", "Tommy"],
    "species": ["Snake", "Giraffe", "Leopard", "Monkey", "Bear", "Panda"],
    "age": [98, 50, 6, 45, 100, 26],
    "weight": [464, 41, 328, 463, 50, 349]
}

animals = pd.DataFrame(data)

findHeavyAnimals(animals)
````
