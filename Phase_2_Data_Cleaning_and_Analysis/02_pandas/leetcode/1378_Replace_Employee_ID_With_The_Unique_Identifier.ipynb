# 1378. Replace Employee ID With The Unique Identifier

**Platform:** LeetCode  
**Difficulty:** Easy  
**Topic:** Pandas – Data Merging / Join  

🔗 Problem Link: https://leetcode.com/problems/replace-employee-id-with-the-unique-identifier/

## 📝 Problem Description

Table: `Employees`

Columns:
- id
- name

Each row represents an employee and their name.

Table: `EmployeeUNI`

Columns:
- id
- unique_id

Each row maps an employee `id` to a `unique_id`.

**Task:**

Show the `unique_id` for each employee.

If an employee **does not have a unique_id**, return **null** instead.

Return a table containing:

- unique_id
- name

## 💡 Key Concepts

1. **Merging DataFrames**
   - Use `pd.merge()` to combine `Employees` and `EmployeeUNI`.

2. **Left Join**
   - Use `how='left'` to keep **all employees**, even if they do not have a matching unique ID.

3. **Handling Missing Values**
   - Employees without a match will automatically receive `NaN` (which represents `NULL`).

4. **Selecting Columns**
   - Return only `unique_id` and `name`.

```python
import pandas as pd

def replace_employee_id(employees: pd.DataFrame, employee_uni: pd.DataFrame) -> pd.DataFrame:
    merged_df = pd.merge(employees, employee_uni, how='left', on='id')
    return merged_df[['unique_id', 'name']]
```

## 🔎 Solution Explanation

1. **Merge the tables**
   - `pd.merge(employees, employee_uni, how='left', on='id')` joins the two tables using `id`.

2. **Use a left join**
   - Ensures all employees appear in the result, even if they do not have a `unique_id`.

3. **Handle missing values**
   - If an employee has no match in `EmployeeUNI`, the `unique_id` will be `NaN` (equivalent to SQL `NULL`).

4. **Select required columns**
   - Only `unique_id` and `name` are returned.

**Time Complexity:** O(n + m)  
**Space Complexity:** O(n)

```python
# Example datasets
employees_data = {
    "id": [1, 7, 11, 90, 3],
    "name": ["Alice", "Bob", "Meir", "Winston", "Jonathan"]
}

employee_uni_data = {
    "id": [3, 11, 90],
    "unique_id": [1, 2, 3]
}

employees = pd.DataFrame(employees_data)
employee_uni = pd.DataFrame(employee_uni_data)

# Run solution
replace_employee_id(employees, employee_uni)
```
