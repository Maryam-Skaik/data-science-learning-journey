# 177. Nth Highest Salary

## 📝 Problem Description

We are given a Pandas DataFrame `Employee` with columns:

- id
- salary

Task:
- Find the **nth highest distinct salary** from the Employee table.
- If there are less than n distinct salaries, return `null`.

Return a DataFrame with a single column:

- `getNthHighestSalary(N)` containing the nth highest salary or null.

## 💡 Key Concepts

1. Remove duplicates: `drop_duplicates()`
2. Sort descending: `sort_values(ascending=False)`
3. Indexing to get nth highest: `.iloc[N-1]`
4. Handle cases where N is larger than number of distinct salaries → return None
5. Output as a single-column DataFrame with dynamic column name.

```python
import pandas as pd

def nth_highest_salary(employee: pd.DataFrame, N: int) -> pd.DataFrame:
    # Edge case: N <= 0
    if N <= 0:
        return pd.DataFrame({f'getNthHighestSalary({N})': [None]})
    
    # Get distinct salaries sorted descending
    salaries = employee['salary'].drop_duplicates().sort_values(ascending=False)
    
    # If N is larger than available distinct salaries
    if N > len(salaries):
        return pd.DataFrame({f'getNthHighestSalary({N})': [None]})
    
    # Return nth highest
    return pd.DataFrame({f'getNthHighestSalary({N})': [salaries.iloc[N-1]]})
```

## 🔎 Solution Explanation

Step 1:
- Remove duplicate salaries using `.drop_duplicates()` to consider distinct salaries only.

Step 2:
- Sort descending to have the highest salary first.

Step 3:
- Check if N is valid:
    - If N > number of distinct salaries, return None
    - Else return the value at index N-1

Step 4:
- Wrap the result in a DataFrame with column name `getNthHighestSalary(N)`.

Time Complexity: O(n log n) for sorting  
Space Complexity: O(n)

```python
data = {
    "id": [1, 2, 3, 4, 5, 6],
    "salary": [100, 200, 200, 300, 400, 700]
}
employee = pd.DataFrame(data)

nth_highest_salary(employee, 2)  # Expected output: 400
```

```python
nth_highest_salary(employee, 5)  # Expected output: 100
```

```python
nth_highest_salary(employee, 6)  # Expected output: None
```

This problem reinforces:

- Deduplication of values with `.drop_duplicates()`
- Sorting in Pandas
- Safe indexing with `.iloc[]`
- Handling dynamic column names in DataFrames
- Returning None/null for invalid positions
