# 178. Rank Scores

**Platform:** LeetCode  
**Difficulty:** Medium  
**Topic:** Pandas – Ranking / Sorting

## 📝 Problem Description

Table: `Scores`

| Column Name | Type    |
|-------------|---------|
| id          | int     |
| score       | decimal |

- `id` is the primary key.
- Each row contains the score of a game (float with two decimal places).

**Task:**  

Rank the scores from highest to lowest:  
1. If there is a tie between two scores, both should have the same rank.  
2. After a tie, the next rank should be the next consecutive integer (no gaps).  
3. Return the table ordered by score in descending order.

**Example Input:**

| id | score |
|----|-------|
| 1  | 3.50  |
| 2  | 3.65  |
| 3  | 4.00  |
| 4  | 3.85  |
| 5  | 4.00  |
| 6  | 3.65  |

**Expected Output:**

| score | rank |
|-------|------|
| 4.00  | 1    |
| 4.00  | 1    |
| 3.85  | 2    |
| 3.65  | 3    |
| 3.65  | 3    |
| 3.50  | 4    |

## 💡 Key Concepts

1. **Ranking**:  
   - Use `rank(method='dense', ascending=False)` to get ranks without gaps.  
2. **Sorting**:  
   - Sort the DataFrame by `score` in descending order.  
3. **Selecting Columns**:  
   - Return only `score` and `rank` columns for the final output.

```python
import pandas as pd

def order_scores(scores: pd.DataFrame) -> pd.DataFrame:
    # Step 1: Dense ranking, highest score gets rank 1
    scores['rank'] = scores['score'].rank(method='dense', ascending=False)
    
    # Step 2: Sort by score descending
    return scores[['score', 'rank']].sort_values(by='score', ascending=False)
```

## 🔎 Solution Explanation

1. **Ranking**:  
   - `rank(method='dense', ascending=False)` ranks the scores from highest to lowest.  
   - `dense` ensures tied scores get the same rank, and the next rank is consecutive.  

2. **Sorting**:  
   - After ranking, sorting by `score` in descending order gives the final output table.

**Time Complexity:** O(n log n) due to sorting  
**Space Complexity:** O(n)  

```python
# Sample Scores DataFrame
data = {
    'id': [1, 2, 3, 4, 5, 6],
    'score': [3.50, 3.65, 4.00, 3.85, 4.00, 3.65]
}

scores = pd.DataFrame(data)

# Running the function
order_scores(scores)
```
