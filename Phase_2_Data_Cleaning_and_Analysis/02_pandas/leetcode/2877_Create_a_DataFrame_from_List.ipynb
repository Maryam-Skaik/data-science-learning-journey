## 📝 Problem Description

We are given a 2D list called `student_data`.  
Each inner list contains:

- `student_id`
- `age`

Our task is to create a **Pandas DataFrame** from this 2D list.

### Requirements:
- The DataFrame must have two columns:
  - `student_id`
  - `age`
- The row order must remain the same as the original list.

### Example

Input:
[
  [1, 15],
  [2, 11],
  [3, 11],
  [4, 20]
]

Output:

| student_id | age |
|------------|-----|
| 1          | 15  |
| 2          | 11  |
| 3          | 11  |
| 4          | 20  |

## 💡 Key Concept

In Pandas, we can create a DataFrame from:

- A dictionary
- A list of dictionaries
- A NumPy array
- A 2D list

To create a DataFrame from a 2D list:

```python
pd.DataFrame(data, columns=[...])
```

- `data` → The 2D list
- `columns` → List of column names

````python
import pandas as pd
from typing import List

def createDataframe(student_data: List[List[int]]) -> pd.DataFrame:
    return pd.DataFrame(student_data, columns=['student_id', 'age'])
````

## 🔎 Solution Explanation

Step 1: Import pandas  
We use pandas to create and manipulate DataFrames.

Step 2: Define the function  
The function receives a 2D list where:
- Each inner list contains [student_id, age]

Step 3: Create the DataFrame  

```python
pd.DataFrame(student_data, columns=['student_id', 'age'])
```

- `student_data` becomes the table data.
- `columns` assigns names to each column.
- Pandas automatically keeps the original order.

Time Complexity: O(n) <br>
Space Complexity: O(n)

````python
student_data = [
    [1, 15],
    [2, 11],
    [3, 11],
    [4, 20]
]

df = createDataframe(student_data)
df
````
