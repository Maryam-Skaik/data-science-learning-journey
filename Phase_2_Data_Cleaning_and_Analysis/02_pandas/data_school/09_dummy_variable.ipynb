# How do I create dummy variables in pandas?

```python
import pandas as pd
```

```python
train = pd.read_csv('https://bit.ly/kaggletrain')
train.head(15)
```

```python
train['sex_male'] = train.Sex.map({'female': 0, 'male': 1})
```

```python
train.head(15)
```

```python
pd.get_dummies(train.Sex) # create a columns to each value
```

```python
pd.get_dummies(train.Sex).iloc[:, 1:]
```

```python
pd.get_dummies(train.Sex, prefix='Sex').iloc[:, 1:]
```

```python
# can we also try it with columns with more than 2 values
train.Embarked.value_counts()
```

```python
pd.get_dummies(train.Embarked, prefix='Embarked')
```

```python
pd.get_dummies(train.Embarked, prefix='Embarked').iloc[:, 1:] # we drop C column
```

```python
embarked_dummies = pd.get_dummies(train.Embarked, prefix='Embarked').iloc[:, 1:]
pd.concat([train, embarked_dummies], axis=1)
```

```python
train = pd.read_csv('https://bit.ly/kaggletrain')
train.head(15)
```

```python
pd.get_dummies(train, columns=['Sex', 'Embarked'])
```

```python
pd.get_dummies(train, columns=['Sex', 'Embarked'], drop_first=True)
```
