# How do I apply a function to a pandas Series or DataFrame?

```python
import pandas as pd
```

```python
train = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
train.head(15)
```

```python
# map function
# if we want to change Sex values to ones and zeros
train['Sex_num'] = train.Sex.map({'female': 0, 'male': 1})
```

```python
train.loc[0:14, ['Sex', 'Sex_num']]
```

```python
# now, move to apply, which is both: Series method and Dataframe method
# if we want to count length of name 
train['Name_length'] = train.Name.apply(len) # here we use it with Series
```

```python
train.loc[0:14, ['Name', 'Name_length']]
```

```python
# another example to round Fare column
import numpy as np
train['Fare_ceil'] = train.Fare.apply(np.ceil)
```

```python
train.loc[0:14, ['Fare','Fare_ceil']]
```

```python
# let's use apply to extract first name from name column
train.Name.str.split(',').head(15) # note that result is list of 2 element, first and first name
```

```python
# create a function to return first name, and try it with apply method
def get_element(my_list, position):
    return my_list[position]
```

```python
train.Name.str.split(',').apply(get_element, position = 0) # now we got first name, since posistion = 0
```

```python
# also we can done this without using function, instead of it use lambda
train.Name.str.split(',').apply(lambda x : x[0]).head(15)
```

```python
# now, we also can use apply method with Dataframe
drinks = pd.read_csv('https://raw.githubusercontent.com/justmarkham/DAT8/master/data/drinks.csv')
drinks.head(15)
```

```python
drinks.loc[:, 'beer_servings':'wine_servings']
```

```python
drinks.loc[:, 'beer_servings':'wine_servings'].apply(max, axis=1)
```

```python
drinks.loc[:, 'beer_servings':'wine_servings'].apply(max, axis=0)
```

```python
drinks.loc[:, 'beer_servings':'wine_servings'].apply(np.argmax, axis=1)
```

```python
# last one is: applymap, applied to every element in df
drinks.loc[:, 'beer_servings':'wine_servings'].applymap(float)
```

```python
# use map instead applymap
drinks.loc[:, 'beer_servings':'wine_servings'].map(float)
```

```python
# we can use this to overwrite df
drinks.loc[:, 'beer_servings':'wine_servings'] = drinks.loc[:, 'beer_servings':'wine_servings'].map(float)
```
