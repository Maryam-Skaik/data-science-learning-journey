# How do I use the "axis" parameter in pandas?

```python
import pandas as pd
```

```python
drinks = pd.read_csv('http://bit.ly/drinksbycountry')
drinks.head(15)
```

```python
# if we want to remove the continent columns
drinks.drop('continent', axis=1).head()
```

```python
# if I change axis to 0, it will reomve row
drinks.drop(2, axis=0).head()
```

```python
drinks.mean(numeric_only=True) # this by default calculate mean for columns, mean axis=0 by default
```

```python
drinks.mean(numeric_only=True, axis=0)
```

```python
drinks.mean(numeric_only=True, axis=1) # to get for rows
```

```python
drinks.mean(numeric_only=True, axis=1).shape # return # of rows
```

```python
drinks.mean(numeric_only=True, axis=0).shape # return # of numeric columns
```
