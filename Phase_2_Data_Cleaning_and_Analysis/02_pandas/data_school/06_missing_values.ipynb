# How do I handle missing values in pandas?

```python
import pandas as pd
```

```python
ufo = pd.read_csv('https://bit.ly/uforeports')
ufo.head(15)
```

```python
# first method isnull to see null values in dataframe
ufo.isnull().tail(20)
```

```python
# as it, we have also notnull method
ufo.notnull().tail(15)
```

```python
# now, how we can benifit from them, first to know number of missing values in each column
ufo.isnull().sum()
```

```python
# now if we want to see the rows with null values from specific column
ufo[ufo.City.isnull()] # so this will return the 25 rows
```

ok, now we see the missing values, but what we must do with them <br>
the common action, is to drop these rows

```python
# check dataframe shape before and after dropping
ufo.shape
```

```python
ufo.dropna(how='any').shape
```

```python
# we can use subset parameter
ufo.dropna(subset=['City', 'Shape Reported'], how='any').shape
```

```python
ufo.dropna(subset=['City', 'Shape Reported'], how='all').shape
```

```python
# another action with missing value, is to fill them
ufo['Shape Reported'].value_counts(dropna=False)
```

```python
ufo['Shape Reported'] = ufo['Shape Reported'].fillna(value='VARIOUS')
```

```python
ufo['Shape Reported'].value_counts(dropna=False)
```
