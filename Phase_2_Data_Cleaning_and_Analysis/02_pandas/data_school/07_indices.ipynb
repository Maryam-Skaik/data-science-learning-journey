# What do I need to know about the pandas index?

```python
import pandas as pd
```

```python
drinks = pd.read_csv('https://bit.ly/drinksbycountry')
```

```python
drinks.head(15)
```

```python
drinks.index # index is numbers
```

```python
# but if we try to see columns, we won't find index column
drinks.columns
```

```python
# even if we try to see shape, we won't see it count index as column
drinks.shape
```

```python
# we can change this index
drinks.set_index('country', inplace=True)
drinks
```

```python
# now, if we try to see columns, we won't see country, since it become an index
drinks.columns
```

```python
# same if we check shape
drinks.shape
```

```python
# if we want to remove country name as the index column name
drinks.index.name = None
drinks
```

```python
# now, if we want to reset index to it's default
drinks.index.name = 'country'
drinks.reset_index(inplace=True)
drinks
```
