# How do I create a pandas DataFrame from another object?

```python
import pandas as pd
```

```python
# first create df from dictionary
pd.DataFrame({'id':[100, 101, 102], 'color':['red', 'blue', 'red']})
```

```python
df = pd.DataFrame({'id':[100, 101, 102], 'color':['red', 'blue', 'red']}, columns=['id', 'color'], index=['a','b','c'])
df
```

```python
# create df from list of lists
pd.DataFrame([[100, 'red'],[101, 'blue'], [102, 'red']])
```

```python
pd.DataFrame([[100, 'red'],[101, 'blue'], [102, 'red']], columns=['id', 'color'])
```

```python
# create df from array of numbers
import numpy as np

arr = np.random.rand(4,2)
arr
```

```python
pd.DataFrame(arr, columns=['one', 'two'])
```

```python
pd.DataFrame({'student': np.arange(100, 110, 1), 'test':np.random.randint(60, 101)})
```

```python
pd.DataFrame({'student': np.arange(100, 110, 1), 'test':np.random.randint(60, 101)}).set_index('student')
```
