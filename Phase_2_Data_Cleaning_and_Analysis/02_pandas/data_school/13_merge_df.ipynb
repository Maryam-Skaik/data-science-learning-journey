# How do I merge DataFrames in pandas?

```python
import pandas as pd
```

## Movies

```python
movie_cols = ['movie_id', 'title']
movies = pd.read_table('u.item', sep='|', header=None, names=movie_cols, usecols=[0,1], encoding='ISO-8859-1')
movies.head()
```

```python
movies.shape
```

```python
movies.movie_id.nunique()
```

## Ratings

```python
rating_cols = ['user_id', 'movie_id', 'rating', 'timestamp']
ratings = pd.read_table('u.data', sep='\t', header=None, names=rating_cols, usecols=[0,1,2,3], encoding='ISO-8859-1')
ratings.head()
```

```python
ratings.shape
```

```python
ratings.movie_id.nunique()
```

```python
ratings.loc[ratings.movie_id == 1, :].head()
```

## Merging Movies and Ratings

```python
# first we must know which columns are in each dataframe
movies.columns
```

```python
ratings.columns
```

```python
movie_ratings = pd.merge(movies, ratings)
movie_ratings.columns
```

```python
movie_ratings.head()
```

```python
movie_ratings.shape
```

## Now, What if the columns you want to join on don't have the same name?

```python
movies.columns = ['m_id', 'title']
movies.columns
```

```python
ratings.columns
```

```python
pd.merge(movies, ratings, left_on='m_id', right_on='movie_id').head()
```

## What if you want to join on one index?

```python
movies = movies.set_index('m_id')
movies.head()
```

```python
pd.merge(movies, ratings, left_index=True, right_on='movie_id').head()
```
