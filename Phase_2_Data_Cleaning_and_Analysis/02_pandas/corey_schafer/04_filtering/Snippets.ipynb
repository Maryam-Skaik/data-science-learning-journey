```python
people = {
    "first": ["Maryam", 'Tia', 'Tima'], 
    "last": ["Skaik", 'lulu', 'lulu'], 
    "email": ["MarymSkaik@gmail.com", 'Tialulu@email.com', 'Timalulu@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['last'] == 'lulu'
```

```python
filt = (df['last'] == 'lulu')
```

```python
df[filt]
```

```python
# we will get same result with loc
df.loc[filt]
```

```python
df.loc[filt, 'email']
```

```python
filt = (df['last'] == 'lulu') & (df['first'] == 'Tia')
```

```python
df.loc[filt, 'email']
```

```python
filt = (df['last'] == 'Skaik') | (df['first'] == 'Tia')
df.loc[filt, 'email']
```

```python
# the opposite 
df.loc[~filt, 'email']
```
