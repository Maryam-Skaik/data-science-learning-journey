# Filtering - Using Conditionals to Filter Rows and Columns

```python
import pandas as pd
```

```python
df = pd.read_csv('data/survey_results_public.csv', index_col='ResponseId')
schema_df = pd.read_csv('data/survey_results_schema.csv', index_col='qname')
```

```python
pd.set_option('display.max_columns', 171)
pd.set_option('display.max_rows', 171)
```

```python
df.head()
```

```python
high_salary = (df['ConvertedCompYearly'] > 80000)
```

```python
df.loc[high_salary]
```

```python
df.loc[high_salary, ['Country', 'LanguageHaveWorkedWith', 'ConvertedCompYearly']]
```

```python
countries = ['United States', 'Ukraine', 'India', 'Brazil', 'Australia']
```

```python
filt = df['Country'].isin(countries)
df.loc[filt, 'Country']
```

```python
df['LanguageHaveWorkedWith']
```

```python
filt = df['LanguageHaveWorkedWith'].str.contains('Python', na=False)
filt
```
