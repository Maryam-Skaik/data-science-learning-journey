```python
import pandas as pd
```

```python
india_df.to_csv('data/modified.tsv', sep='\t')
```

```python
# read and write SQL
import sqlite3
import pandas as pd
```

```python
conn = sqlite3.connect("data/analysis.db")
```

```python
india_df.to_sql(
    "sample_table",
    conn,
    if_exists="replace",
    index=False
)
```

```python
df = pd.read_sql(
    "SELECT * FROM sample_table LIMIT 10",
    conn
)
df
```

```python
sql_df = pd.read_sql_query(
    "SELECT * FROM sample_table",
    conn,
    index_col="MainBranch"
)
sql_df
```
