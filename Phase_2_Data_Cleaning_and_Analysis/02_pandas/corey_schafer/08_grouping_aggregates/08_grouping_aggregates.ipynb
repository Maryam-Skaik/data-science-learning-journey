```python
# to get a percentage
df['CommPlatformHaveEntr'].value_counts(normalize=True)
```

```python
df['Country'].value_counts()
```

```python
# to group result by country
df.groupby(['Country']) # we got DataFrameGroupBy object
```

```python
# assign to a variable
country_grp = df.groupby(['Country'])
country_grp.get_group('Germany') # we see it return a dataframe
```

```python
# as we did before to got coutries using filter, we will face same result
filt = df['Country'] == 'Germany'
df.loc[filt]
```

```python
# now if we want the most platforms using in this country using filter
filt = df['Country'] == 'Germany'
df.loc[filt]['CommPlatformHaveEntr'].value_counts()
```

```python
# here we can see how to benifit from grouping, where groupby country directly help me to find these results
country_grp['CommPlatformHaveEntr'].value_counts().head(50)
```

```python
# same to got the result for scecific country
country_grp['CommPlatformHaveEntr'].value_counts().loc['India']
```

```python
country_grp['CommPlatformHaveEntr'].value_counts(normalize=True).loc['India']
```

```python
# same if we want to see the median salary for each country
country_grp['ConvertedCompYearly'].median()
```

```python
# to see for specific country, we easily use loc
country_grp['ConvertedCompYearly'].median().loc['Germany']
```

```python
country_grp['ConvertedCompYearly'].agg(['median', 'mean'])
```

```python
country_grp['ConvertedCompYearly'].agg(['median', 'mean']).loc['Canada']
```

```python
country_grp['LanguageHaveWorkedWith'].str.contains('Python').sum() # we can't treat with SeriesGroupBy as series
```

```python
# here applay method will help
country_grp['LanguageHaveWorkedWith'].apply(lambda x: x.str.contains('Python').sum())
```
