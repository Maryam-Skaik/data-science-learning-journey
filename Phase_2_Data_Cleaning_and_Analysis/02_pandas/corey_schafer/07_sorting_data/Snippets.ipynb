```python
people = {
    "first": ["Maryam", 'Tia', 'Tima', 'Sara', 'lulu'], 
    "last": ["Skaik", 'lulu', 'lulu', 'ahmed', 'lulu'], 
    "email": ["MarymSkaik@gmail.com", 'Tialulu@email.com', 'Timalulu@email.com', 'SaraAhmed@email.com', 'lulu@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
# to sort df we can use sort_values function, and secify the column to sort by
df.sort_values(by='last') # note for maryam -> S < a, since S is upper case, while a is lower case, and upper come before lower
```

```python
# sort desc
df.sort_values(by='last', ascending=False) # now skaik is last one ):
```

```python
# also we can sort by more than one field using list 
df.sort_values(by=['last', 'first'], ascending=False)
```

```python
# now if we want to sort by first col as asc, and other as desc, so make list also for ordering
df.sort_values(by=['last', 'first'], ascending=[True,False])
```

```python
# applay to our df
df.sort_values(by=['last', 'first'], ascending=[True,False], inplace=True)
df
```

```python
# now to resort index
df.sort_index()
```
