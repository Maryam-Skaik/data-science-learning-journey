```python
people = {
    "first": ["Maryam", 'Tia', 'Tima'], 
    "last": ["Skaik", 'lulu', 'lulu'], 
    "email": ["MarymSkaik@gmail.com", 'Tialulu@email.com', 'Timalulu@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df.columns
```

```python
# we can change columns names by assign new values to columns
df.columns = ['first_name', 'last_name', 'email']
df
```

```python
df.columns
```

```python
# now if we want to change names from upper case to lower case and vice versa
# we can do this with list comprehension
df.columns = [x.upper() for x in df.columns]
df
```

```python
# if we want to replace _ with space
df.columns = df.columns.str.replace('_', ' ')
df
```

```python
# because we need names with _, we will make replace again, and also convert it to lower case
df.columns = df.columns.str.replace(' ', '_')
df.columns = [x.lower() for x in df.columns]
df
```

```python
# another way to rename columns, is treat with it as dictionary, key is old name, and value is the new
df.rename(columns={'first_name': 'first', 'last_name': 'last'})
```

```python
# but dataframe still has the old names, so we have to add an inplace parameter
df
```

```python
# fix
df.rename(columns={'first_name': 'first', 'last_name': 'last'}, inplace=True)
df
```

```python
# now we updates columns names, but what if we want to update the data in the rows ?
# fisrt way as assignment with loc, mean get the index of row, and columns name then change the cell value
df.loc[2, 'last'] = 'el-lulu'
df
```

```python
# same we can change the whole row data
df.loc[2] = ['saly', 'ahmed', 'salyahmed@email.com']
df
```

```python
# we can do same with at instead of loc
df.loc[2, ['first','last']] = ['Tima', 'lulu']
df
```

```python
filt = (df['email'] == 'salyahmed@email.com')
df.loc[filt, 'email'] = 'Timalulu@email.com'
df
```

```python
# we updated single row of data, now we want to update multiple rows of data
df['email'] = df['email'].str.lower()
df['email']
```

```python
df
```

```python
# using applay function
df['email'].apply(len)
```

```python
# we can benifit from it, if we have a function, and want to applay to data
def update_email(email):
    return email.upper()
```

```python
df.email.apply(update_email)
```

```python
df.email = df.email.apply(update_email)
df
```

```python
# now return it to lower case using applay and lambda
df.email = df.email.apply(lambda x : x.lower())
df
```

```python
df.apply(len) # return # of values to each columns
```

```python
# same as:
df.apply(len, axis='columns')
```

```python
# to get max from each column
df.apply(pd.Series.max)
```

```python
df.apply(lambda x: x.min())
```
