```python
import pandas as pd
import numpy as np
```

```python
people = {
    'first': ['Maryam', 'Tia', 'Tima', 'Sara', np.nan, None, 'NA'], 
    'last': ['Skaik', 'lulu', 'lulu', 'ahmed', np.nan, np.nan, 'Missing'], 
    'email': ['MaryamSkaik@gmail.com', 'Tialulu@email.com', 'Timalulu@email.com', None, np.nan, 'Anonymous@email.com', 'NA'],
    'age': ['22', '20', '20', '25', None, None, 'Missing']
}
```

```python
df = pd.DataFrame(people)
df
```

```python
df.dropna() # drop all None
```

```python
df.dropna(axis='index', how='any') # give same result
# index mean drop rows contains missing values, but if we put columns, then it will drop the whole column contain a missing value
# and for how, this specify how to drop rows, we choose drop row if any of it's value is missing
```

```python
# if we try how=all
df.dropna(axis='index', how='all') # just drop rows with all None or NaN
```

```python
df.dropna(axis='columns', how='all') # same df, since no columns contains all missing values
```

```python
df.dropna(axis='columns', how='any') # give empty df
```

```python
# we ca specify which columns to focus on for missing values, using subset paramerer
df.dropna(axis='index', how='any', subset=['email']) # cheack only one value, even with how=all
```

```python
df.dropna(axis='index', how='all', subset=['email']) # same result
```

```python
# to specify more than one columns to check
df.dropna(axis='index', how='all', subset=['last', 'email']) # mean must email and last be missing to drop 
```

```python
# ok, now want to solve these values with string = NA, or missing, so we can drop easily
df.replace('NA', np.nan, inplace=True)
df.replace('Missing', np.nan, inplace=True)
df
```

```python
# now we can run df.dropna
df.dropna()
```

```python
# to see missing values as T, F
df.isna()
```

```python
# if we process missing values with different way, mean we don't want to drop the whole row contains a missing value
# we can fill it with value we want
df.fillna("MISSING")
```

```python
# or we can fill with zeros, if we care about numbers, like age, score, gpa 😏
df.fillna(0)
```

```python
# get mean of ages, but check dt before
df.dtypes
```

```python
# we find age is object, so take the mean directly will cause a problem
df['age'].mean()
```

```python
# first change age data type
df['age'] = df['age'].astype(float)
df.dtypes
```

```python
# now we can safely get the mean
df['age'].mean()
```
