```python
people = {
    "first": ["Maryam", 'Tia', 'Tima'], 
    "last": ["Skaik", 'lulu', 'lulu'], 
    "email": ["MarymSkaik@gmail.com", 'Tialulu@email.com', 'Timalulu@email.com']
}
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(people)
```

```python
df
```

```python
df['first'] + ' ' + df['last']
```

```python
# to add a new column, we just make assignment
df['full_name'] = df['first'] + ' ' + df['last']
df
```

```python
# to reomve or delete a column, we use drop method
df.drop(columns=['first', 'last'])
```

```python
# same if we see df, it will be as it, to change it add inplace=True
```

```python
df.drop(columns=['first', 'last'], inplace=True)
df
```

```python
# to return columns we deleted by using full_name
df['full_name'].str.split(' ', expand=True)
```

```python
# now we just give them names, and assign them
df[['first', 'last']] = df['full_name'].str.split(' ', expand=True)
df
```

```python
# to add new row
df = pd.concat([df, pd.DataFrame([{'first': 'Saly'}])])
df
```

```python
people = {
    'first': ['Sara', 'Lamar'], 
    'last': ['Skaik', 'Skaik'], 
    'email': ['SaraSkaik@email.com', 'LamarSkaik@email.com']
}
df2 = pd.DataFrame(people)
df2
```

```python
# now we want to add df2 to df
df = pd.concat([df, df2], ignore_index=True)
df
```

```python
# to drop using index
df.drop(index=3, inplace=True)
df
```

```python
filt = df['last'] == 'lulu'
df.drop(index=df[filt].index)
```
