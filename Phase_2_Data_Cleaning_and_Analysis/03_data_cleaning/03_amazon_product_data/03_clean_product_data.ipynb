# Data Cleaning with Python Pandas: Step-by-Step Guide on Amazon Product Data!

```python
import pandas as pd
```

```python
data = pd.read_csv('./amazon_product.csv')
data
```

```python
data.columns
```

```python
data.info()
```

```python
data.drop(columns= ['Unnamed: 0', 'asin', 'product_url', 'product_photo', 'product_num_offers', 'product_minimum_offer_price', 'is_amazon_choice', 'is_prime', 'climate_pledge_friendly',
       'sales_volume', 'delivery', 'has_variations'], inplace=True)
```

```python
data.head(10)
```

```python
data_new_columns_names = []

for i in data.columns:
    data_new_columns_names.append(i.capitalize())
```

```python
data_new_columns_names
```

```python
data.columns = data_new_columns_names
```

```python
data.head(1)
```

```python
data.duplicated().sum()
```

```python
data.isna().sum()
```

```python
data.isna().sum().sum()
```

Note these columns with the most missing values, so if we make data.dropna() we lose the dataset <br>
Product_availability      63 <br>
Unit_price                59 <br>
Unit_count                59 <br>


So, we can just remove these coloumns

```python
data = data[['Product_title', 'Product_price', 'Currency', 'Product_star_rating', 'Product_num_ratings', 'Is_best_seller']]
```

```python
data
```

```python
data.isna().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isna().sum()
```

```python
data.loc[:, 'Product_price'] = data['Product_price'].str.replace('$', '', regex=False)
```

```python
data
```
